## 1. O que é um módulo? E um pacote ou biblioteca?

- **Módulo:** ficheiro Python (`.py`) que contém código reutilizável, como funções, classes e constantes.
- **Pacote:** estrutura que agrupa módulos, normalmente num diretório. Também existem *namespace packages*.
- **Biblioteca:** designação geral para um conjunto de ferramentas reutilizáveis, que pode incluir vários módulos e pacotes.
- **Biblioteca padrão:** ferramentas distribuídas com Python, como `math`, `datetime` ou `json` (sem instalação adicional).
- **Biblioteca externa:** pode necessitar de instalação, como `pandas`, `numpy` ou `requests`.

Uma **distribuição** instalada com `pip` nem sempre tem o mesmo nome que o pacote importado: instalamos `beautifulsoup4`, mas importamos `bs4`.

O código pode ser dividido em vários módulos para facilitar a organização, a manutenção e a reutilização.

### Formas de importar

- `import math` - acesso através de `math.sqrt(9)`.
- `import math as m` - cria um nome alternativo.
- `from math import sqrt` - importa apenas um nome.

Evita `from modulo import *`: torna menos evidente a origem dos nomes.

````python
import math
import math as m
from datetime import date

print(math.sqrt(81))
print(m.pi)
print(date.today())
````

### `sqlite3` - Base de dados SQLite em memória

📚 [Documentação oficial](https://docs.python.org/3/library/sqlite3.html)

````python
import sqlite3
with sqlite3.connect(':memory:') as ligacao:
    ligacao.execute('CREATE TABLE alunos (nome TEXT, nota INTEGER)')
    ligacao.executemany('INSERT INTO alunos VALUES (?, ?)', [('Ana', 18), ('Rui', 16)])
    linhas = ligacao.execute('SELECT nome, nota FROM alunos ORDER BY nota DESC').fetchall()
    print(linhas)
````

**Experimenta:** Explica por que razão `?` é usado para inserir valores em SQL.

### `unittest` - Testes automáticos de funções

📚 [Documentação oficial](https://docs.python.org/3/library/unittest.html)

````python
import unittest

def dobro(numero):
    return numero * 2

class TestarDobro(unittest.TestCase):
    def test_dobro(self):
        self.assertEqual(dobro(4), 8)

resultado = unittest.TestResult()
TestarDobro('test_dobro').run(resultado)
print('Testes executados:', resultado.testsRun, '| Erros:', len(resultado.errors), '| Falhas:', len(resultado.failures))
````


## 8. Miniprojeto integrador - Registo de notas

Este projeto utiliza `csv`, `pathlib`, `statistics`, `json` e `collections`, sem instalações adicionais. Reúne as técnicas aprendidas em diferentes módulos.


````python
from pathlib import Path
from collections import Counter
import csv
import json
import statistics

registos = [('Ana', 18), ('Rui', 16), ('Marta', 18), ('Paulo', 14)]
ficheiro = Path('projeto_notas_cap20.csv')
with ficheiro.open('w', newline='', encoding='utf-8') as f:
    escritor = csv.writer(f)
    escritor.writerow(['Nome', 'Nota'])
    escritor.writerows(registos)

with ficheiro.open('r', newline='', encoding='utf-8') as f:
    linhas = list(csv.DictReader(f))

notas = [int(linha['Nota']) for linha in linhas]
resumo = {
    'numero_alunos': len(notas),
    'media': round(statistics.mean(notas), 2),
    'nota_mais_frequente': Counter(notas).most_common(1)[0][0]
}
Path('projeto_resumo_cap20.json').write_text(
    json.dumps(resumo, indent=2, ensure_ascii=False), encoding='utf-8'
)
print('Resumo:', resumo)
print('Ficheiros criados:', ficheiro.name, 'e projeto_resumo_cap20.json')
````
