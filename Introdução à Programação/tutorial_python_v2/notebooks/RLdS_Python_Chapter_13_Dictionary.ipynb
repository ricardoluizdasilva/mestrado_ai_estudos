# Dicionários

Ao contrário das seqüências, que são indexadas por um intervalo de números, os `dicionários são indexados por chaves`, que podem ser de qualquer tipo imutável. É uma coleção não ordenada de pares de chave e valor, permitem trabalhar com grande quantidade de dados e são otimizados para buscar dados através da sua chave.

Um dicionário é definido entre chaves `{}` onde cada item é um par na forma de `chave:valor`, separados por vírgula `,`, os quais podem ser de qualquer tipo.

Características das chaves:
- Devem ser únicas;
- Imutável (int, float, string, tuple, bool);
- Nenhuma ordem para chaves ou valores.

Características dos valores:
- Qualquer tipo de dados (imutável ou mutável);
- Pode ser duplicado;
- Valores de dicionários podem ser listas, ou até mesmo outros dicionários.

Alguns exemplos para entender melhor o funcionamento dos dicionários.



```python
# Agenda telefónica
agenda = {
    "Luke Skywalker": 11223344,
    "Dart Vader": 666666,
    "Obi-Wan Kenobi": 98765432,
    "Han Solo": 9998873,
}
print(agenda)
```

```python
# Registo de aluno

aluno = {
    "sobrenome": "Skywalker",
    "nome": "Anakin",
    "posicao": "Padawan",
    "matriculado": True,
    "habilidades": ["piloto", "eletrónica", "mecânica"],
    "disciplinas": ["Sabre de Luz", "Pilotagem"],
}
print(aluno)
```

Novamente, o conhecido método `len()` permite verificar o **tamanho de um dicionário**.

```python
print("O tamanho do dicionário agenda é", len(agenda))
print("O tamanho do dicionário aluno é", len(aluno))
```

## Aceder aos elementos de um dicionário

Para aceder aos elementos de uma `list` e uma `tuple` utilizamos o índice.

Isto tem algumas desvantagens quando se quer aceder a um elemento específico, pois é preciso saber qual é o seu respetivo índice. E isto pode ficar ainda mais complicado quando há uma enorme quantidade de dados e elementos são constantemente incluídos ou removidos, o que muda a posição do índice do elemento, tornando a programação muito mais complexa.

Contudo, este problema pode ser resolvido com *dicionários*. Assim com as lists ou tuples, os dicionários podem gerir múltiplos elementos. A grande diferença é que os elementos são acedidos pelos seus nomes (chaves) e não pelos índices.

Vejamos o exemplo da lista telefónica:

| Nome            | Telefone         |
| :-------------- | ---------------: |
| Darth Vader     |          666 666 |
| Obi-Wan Kenobi  |       987 654 32 |
| Luke Skywalker  |      11 22 33 44 |
| Han Solo        |        999 88 73 |


Neste exemplo, a pesquisa **não** será feito por um terceiro elemento como o `índice`, mas através do número de telefone que pertença ao nome pesquisado. 

Como mencionado anteriormente, um dicionário consiste de pares de valores e chaves, onde cada valor-chave é representado por `key: value`. Um dicionário poder ter vários pares, como `{key1: value1, key2: value2, ..., keyN: valueN}`.

Vejamos na **célula abaixo** como aceder aos elementos de um dicionário.


```python
# Pesquisar o telefone de Obi-Wan kenobi
print("O telefone de Obi-Wan Kenobi é", agenda["Obi-Wan Kenobi"])

# Pesquisar as disciplinas de Anakin
print("As disciplinas de Anakin são", aluno["disciplinas"])
```


O método `get()` permite também aceder aceder ao elemento do dicionário.


```python
print(aluno.get("nome"))
```


A `função keys()` permite obter todas as chaves do dicionário.


```python
print(aluno.keys())
```


A utilização do `for` permite **iterar sobre os elementos de um dicionário**.


```python
for chave in aluno:
    print(aluno[chave])
```


Outra **alternativa** seria utilizar o `método values()`.


```python
for valor in aluno.values():
    print(valor)
```


Além disto é possível **percorrer** tanto as *chaves* como os *valores*. Para tal utiliza-se a `função items()`.


```python
for chave, valor in aluno.items():
    print(f"{chave} : {valor}")
```

A tabela abaixo sumariza as **funções/métodos** exemplificados acima.

| Função/Método   | Valor de retorno                            |
| --------------- | ------------------------------------------- |
| `len()`         | Número de pares `key:value` no dicionário   |
| `.keys()`       | Todas as chaves do dicionário               |
| `.values()`     | Todos os valores do dicionário              |
| `.items()`      | Todos os pares `key:value` como tuples      |

> **Nota:** o tipo de dados de saída do método `.keys()` é `dict_keys`. Com a ajuda da função `list()`, isto pode ser convertido em uma lista.


```python
print(agenda.keys())
print(list(agenda.keys()))
```

```python
print(agenda.values())
print(list(agenda.values()))
```

```python
print(agenda.items())
print(list(agenda.items()))
```

A utilização do operador `in` permite verificar se um determinado item **está presente** no dicionário. E o `operador inverso` também pode ser utilizado.

```python
print("nome" in aluno)
print("Anakin" in aluno.values())
print("Han Solo" not in agenda)
print("Yoda" in agenda.keys())
print(666666 in agenda.values())
```

## Operações básicas

### `Adicionar um elemento ao dicionário com input()`

Podemos utilizar a função `input()` para adicionar valores a um dicionário conforme exemplo da **célula abaixo**.


```python
nome = input("Digite o seu nome: ")
fone = input("Digite o seu telefone: ")

agenda[nome] = fone

print(agenda)
```

### `Ordenar um dicionário`

É possível utilizar o `método sorted()` para **ordenar um dicionário**. A ordenação será feita na utilização em conjunto do `método items()`; retornará uma lista de `tuple`.

**Exemplo de ordenação crescente:**


```python
print(sorted(agenda.items()))
```

**Exemplo de ordenação decrescente:**

```python
print(sorted(agenda.items(), reverse=True))
```

### `Remover itens do dicionário`

É possível remover itens de um dicionários.

**Por exemplo**, o `método pop()` é utilizado para remover um elemento; retorna o item removido.


```python
agenda.pop('Ricardo')
```

O `método opitem()` remove o último par de chave-valores adidionado ao dicionári; retorna a chave-valor removida.

```python
agenda.popitem()
```

A `palavra-chave del` remove um item com a chave especificada.

```python
del aluno["habilidades"]
print(aluno)
```

Ou ainda, usar `del` para apagar o dicionário por completo, ou seja, a variável é removida da memória.

```python
del aluno
```

Já o `método clear()` permite limpar o dicionário por completo. Neste caso, a variável permanece na memória, mas sem os pares chaves-valores.

```python
agenda.clear()
print(agenda)
```

## Dicionário comprehensions

Podemos usar a técnica de **comprehensions** para `criar um dicionário` a partir de duas listas, por exemplo.


```python
# Listas
diretores = ["John Badham", "Phil Alden Robinson", "Iain Softley", "Sam Esmail", "Michael Mann", "Niels Arden Oplev",
             "Lana Wachowski, Lilly Wachowski" ]

filmes = ["WarGames", "Sneakers", "Hackers", "Mr. Robot", "Blackhat", 
          "The Girl with the Dragon Tattoo", "Matrix"]

# Criar dicionário a partir das listas
diretor_filme = {diretor: filme for diretor, filme in zip(diretores, filmes)}

print(diretor_filme)
```


Um `dicionário comprehension` pode também conter a `instrução if`, que pode ser usada para `filtrar itens`, conforme **exemplo** abaixo onde temos apenas os números pares e os seus respetivos quadrados perfeitos.


```python
quadrados_pares = {x: x**2 for x in range(11) if x % 2 == 0}
print(quadrados_pares) # {0: 0, 2: 4, 4: 16, 6: 36, 8: 64, 10: 100}
```

## Dicionário dentro de outro dicionário

Os dicionários são uma estrutura de dados muito flexível. Permite até mesmo armazenar dicionários dentro de um dicionário, de forma que possam ser acedidos através de uma chave.

Exemplo:


```python
filmes = {
'1':{"filme":"WarGames","direcao":"John Badham"},
'2':{"filme":"The Girl with the Dragon Tattoo","direcao":"Niels Arden Oplev"},
'3':{"filme":"Matrix","direcao":"Lana Wachowski, Lilly Wachowski"}
}

print(filmes['2'])
print(filmes.get('1'))
print(filmes['3']['filme'])
```

## Cópia de dicionários

O `método copy()` permite copiar um dicionário.

> **Nota:** não é possível copiar um dicionário apenas com a sintaxe `var_copia = var`, porque será apenas uma referência para a variável de origem, e as alterações feitas nesta serão automaticamente refletidas em var_copia. Para evitar isto, utiliza-se o método acima ou a alternativa `var_copia = dict(var)`


```python
# método copy()
filme = {'titulo':'Episode IV – A New Hope','Gênero':'Ficção Científica'}
print(filme)
copia_filme_1 = filme.copy()
print(copia_filme_1)
```

```python
copia_filme_2 = dict(filme)
print(copia_filme_2)
filme["titulo"] = "Episode VIII – The Last Jedi"
print(filme)
print(copia_filme_2)
```


## Técnica de memoização

A técnica de memoização `(memoization)` permite otimizar e acelerar programas através do armazenamento do resultado das chamadas de funções complexas que demandam poder de processamento e memória, como por exemplo as funções recursivas.

Para tal podemos utilizar dicionários, pois são estruturas de dados eficazes ao guardar e pesquisar dados.
