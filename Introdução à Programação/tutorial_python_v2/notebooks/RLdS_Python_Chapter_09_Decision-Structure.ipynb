# Estrutura de decisão

Até o presente momento vimos instruções de código sendo executadas uma a uma até que o programa termine. Contudo, é frequente que partes do programa só sejam executadas de acordo com uma determinada condição.

Imagine a seguinte situação: `Amanhã, Iremos à praia se não chover. Caso contrário, iremos ao centro comercial fazer compras e almoçar. A noite iremos jantar com amigos`.

A descrição do texto acima não é totalmete precisa, assim com as tarefas que serão executadas. Esta é uma descrição típica em linguagem natural. É frequente haver necessidade de informação adicional e mais contextualizda, a qual poderá ser obtida através de uma conversa, por **exemplo**. Uma conversão para pseudocodigo ficaria assim:


```Python
Se chover amanhã chover:
    - Ir ao centro comercial
    - Fazer compras
    - Almoçar
Caso contrário:
    - Iremos à praia

A noite iremos jantar com amigos.
```

A **tomada de decisão** é necessária quando se pretente executar parte de um código apenas se uma `determinada condição for satisfeita`. Este é um conceito muito importante, pois representa a capacidade lógica e analítica na execução de instruções.





É importante relembrar o conceito dos operadores relacionais que são utilizados para comparar valores (de qualquer tipo). O resultado de uma expressão relacional será um resultado `booleano` (`True` ou `False`).

| Símbolo   | Descrição        | Exemplo                |Significado                       |
| --------- | ---------------- | ---------------------- |--------------------------------- |
| >         | Maior que        | X > Y                  |`X` é maior que `Y` ?             |
| >=        | Maior ou igual   | X >= Y                 |`X` é maior que ou igual a `Y` ?  |
| <         | Menor            | X < Y                  |`X` é menor que `Y` ?             |  
| <=        | Menor ou igual   | X <= Y                 |`X` é menor que ou igual a `Y` ?  |
| ==        | Igual            | X == Y                 |`X` é igual a `Y` ?               |
| !=        | Diferente        | X != Y                 |`X` é diferente de `Y` ?          | 

É imporante relembrar que o sinal `=` é utilizado para assinalar um valor à uma variável, enquanto o sinal `==` é usado para fazer comparações de igualdade, que retornará `True` se os valores forem iguais ou `False` se forem diferentes.

Essas `condições` podem ser utilizadas de várias maneiras, e são comumente utilizadas em instruções `condicionais (if)` e `loops`.


## Condição básica

Em Python as instruções `if`, `elif` e `else` são usadas para auxiliar na programação da tomada de decisões. Abaixo a representação da sintaxe:

```Python
if <condição_1>:
   instruções_1
   ...
elif <condição_2>:
   instruções_2
   ...
else:
   instruções_3
   ...
```

So indentation and spaces have a meaning in Python! Therefore, you must not indent arbitrarily within a program. Execute the code in the following two cells to see what happens.

Neste exempo, o programa executa a primeira linha `if <condição_1>.` Se a expressão for avalida como `True`, as instruções dentro do `bloco if` *`(instruções_1)`* serão `executadas`.

Caso a expressão da primeira linha retorne `False`, o programa irá passar para a linha `elif <condição_2>`. Se for `True`, as instruções do bloco `elfi` *`(instruções_2)`* serão executadas. Se for falso, as instruções do blolo `else` *`(instruções_3)`* serão executadas.

É importate relembrar que a **`identação`** define os limites dos blocos de código. As instruções que estão juntas são chamadas de blocos de cógido. Como se pode observar anteriormente, Python não permite que carateres especiais ou palavras-chave marquem o início e fim de um bloco de código. Para tal, a identação deve ser utilizada.

Vejamos alguns **exemplos práticos** com `if-elfi-else`.

````python
# Exemplo prático de if-elif-else
# Entrada de dados
x = int(input("Digite o 1º número: "))
y = int(input("Digite o 2º número: "))

# Testa as diferentes condições
if x > y:
    print("O 1º número é maior do que o 2º")
elif x == y:
    print("Ambos os números são iguais")
else:
    print("o 2º número é maior do que o 1º")

````

No `exemplo acima` a comparação é feita com números inteiros. O exemplo pode ser extendido a outros tipos de dados.

Como por **exemplo**, comparação entre `strings`.

````python
# Exemplo prático de if-elif-else
# String
nome = "Darth Vader"
if nome == "Darth Vader":
    print("I am your father")
else:
    print("I am not your father!")
````

Assim como em lógica, em Python (e outras linguagens de programação), a condição pode ser feita pela negação, ou seja, pelo uso do not. Neste caso tem que se observar alguns aspetos importantes:
- `strings vazia` é avalida como `False`. Quaisquer outros valores são considerados `True`.
- `0` é avaliado como `False`. Quaisquer outros números são considerados `True`.

**Exemplos:**


````python
# O comando print("Senha inexistente") será sempre executado uma vez que a string vazia 
senha = ""
if not senha:
    print("Senha inexistente!")
else:
    print("A senha existe!")

# O comando print("Número inexistente!" será sempre executado uma vez que 0 é avaliado como False)
num = 0
if not num:
    print("Número inexistente!")
else:
    print("O número existe!")
````

As **`premissas`** acima podem ser validas através das seguintes instruções em Python:


````python
print(bool(""))
print(bool("A-2msg-493xxx"))
print(bool(0))
print(bool(5))
print(bool(-8))
print(bool(3.38))
print(bool(5+6j))
````

## Múltiplas condições

O que acontece quando há mais de duas alternativas? Por exemplo, se a variável nome contém os valores *João* ou *Maria* ou *António*?

Uma condição só pode ter dois resultados, `True` ou `False`. Contudo, é possível verificar várias condições uma após a outra.

```python
if condição_1:
   instruções_1
   ...
elif condição_2:
   instruções_2
   ...
elif condição_3:
   instruções_3
   ...
elif condição_4:
   instruções_4
   ...
else:
   instruções_5
   ...
```

Neste caso, as condições são verificadas em ordem, uma após a outra. Se a primeira condição for `True`, então o respetivo bloco de código será executado. Caso contrário, passará para a próxima condição, que será validada e os respetivos códigos executado de acordo com o resultado desa validação. Se nenhuma das condições for satisfeita, então o código de bloco do `else`será execuado, ou seja, o bloco de código do `else` só será executado se nenhuma das condições anteriores não for `True`. Se um código de bloco é executado, então nenhuma outra instrução condicional será verificada.

Nas múltiplas condições a ordem das condições é crucial na lógica do programa.

**Exemplo:**

## Expressão condicional
Python também suporta outra entidade de estrutura de decisão, a expressão condicional (conditinal expression, conditional operator ou ainda ternary operator). Ela é caraterizada pela seguinte sintaxe.

```python
<expr1> if <condição_expr> else <expr2>:
```

A diferençã para a instrução if regular acontece porque a expressão acima não controla diretamente o fluxo da execução do programa. Atua mais como um operador que define uma expressão. Na sintaxe acima, `<condição_exp>` é analisada primeiro. Se for `True`, executa então a `<expr1>`. Se for falso, executa `<expr2>`.

É importante realçar segue uma ordem não muito óbvia: a expressão do meio é a primeira a ser avaliada, e com base no resultado, uma das outras expressões será executada terminando assim a condição.

Mais um **exemplo** de uma `expressão condicional`.


````python
idade = 17
print("Menor de idade bebe leite!" if idade < 18 else "Maior de idade pode beber cerveja!")
````
