# Números

Os números podem ser classificados de acordo com um conjunto de números, que vem a ser uma coleção de elementos, mostrados a seguir na representação gráfica.



**Fonte:** https://www.centralexatas.com.br/matematica/conjuntos-numericos/formulas

Em Python existem três tipos de dados numéricos:
- int
- float
- complex

Tudo, em Python, é um objeto, portanto, tipos de dados são classes e variáveis são instâncias (objetos) dessas classes. No decorrer do curso iremos abordar estes aspectos em mais detalhe.

Lembrem-se, os números e suas bases são elementos fundamentais em programação!

Exemplos de variáveis do tipo numérico:
```python
x = 18   # int
y = 18.3 # float
z = 3j   # complex
```

Vejamos na **célula abaixo** o que o interpretador retornará:

````python
x = 18   # int
y = 18.3 # float
z = 3j   # complex
print("x =", x, "é do tipo", type(x))
print("y =", y, "é do tipo", type(y))
print("z =", z, "é do tipo", type(z))
````

### `Definições dos tipos de dados numéricos`

**`- int():`** são números positivos ou negativos e não contêm casas decimais. O seu tamanho está limitado apenas pela capacidade de memória disponível.

```python
x = 5
y = 342907249723902
z = -135
```

**`- float():`** são números positivos ou negativos que podem conter uma ou mais casas decimais. 

```python
x = 5.33
y = 30.0
z = -13.88
```

Para o `float` podemos ainda acrescentar a notação científica `e` ou `E` seguido por um número inteiro positivo ou negativo.


````python
x = 28e3
print(type(x))
print(x)

y = 3.8e-3
print(type(y))
print(y)
````

O valor máximo que o `float` pode ter é de aproximadamente **`1.8x10^308`**. Qualquer número maior que este valor é indicado pelo valor `inf(infinity)`.


````python
print(1.79e308)
print(1.8e308)
````

**`- complex:`** são escritos com `j`representando a parte imaginária. Podem ser escritos `complex(5,6)` ou `5 + 6j`. Um número complexo Python `c` é armazenado internamente usando as `coordenadas cartesianas ou retangulares`.

**Exemplos:**


````python
x = 5+8j
y = -9j
z = complex(5,6)
print(type(x))
print(type(y))
print(type(z))
````

Conforme exemplificado abaixo, podemos ainda usar os métodos `real` e `imag` para obtermos as partes `real` e `imaginária`, respetivamente.


````python
z = complex(5,6)
print(z.real)
print(z.imag)
````

A função abs() retorna a extensão do número complexo, ou seja, representa o cálculo da seguinte fórmula:



````python
z = complex(5,6)
print(abs(z))
````

### `Conversão de números`

Em Python é possível fazer a conversão entre números. Contudo, é importante ressaltar que não é possível converter um número complexo para outro tipo numérico.

**Exemplos:**


````python
x = 6
y = 8.9
z = 5j

print("Conversão de int para float", float(x))
print("Conversão de float para int", int(y))
print("Conversão de int para complex", complex(x))

````

### `Bases`

Normalmente escrevemos os números inteiros na base 10. Entretanto, o Python permite escrever inteiros nos formatos `exadecimal (base 16)`, `octal (base 8)` e `binário (base 2)`. Para tal é preciso adicionar um prefixo ao número inteiro:



**Exemplos:**


````python
print(0b0111111) # 63
print(0o10) # 8
print(0XFF) # 255
````

Os **métodos** abaixo permitem fazer **conversões** entre bases.

- `bin():` converte e retorna a string binária equivalente de um número inteiro.
- `oct():` converte e retorna a string octal de um número inteiro.
- `bin():` converte e retorna a string hexadecimal equivalente de um número inteiro.
- `int():` converte um número para inteiro. Para converter um número de uma determinada base para inteiro é preciso indicar como argumento a base (2, 8 ou 16).

**Exemplos:**

````python
print(bin(255)) # converte 255 em binário: 0b11111111
print(int('0b11111111', 2)) # converte o número binário 0b11111111 em inteiro
print(oct(8)) # converte 8 em octal: 0o10
print(int('0o10',8)) # converte 8 em número octal 0o10 em inteiro
print(hex(255)) # converte 255 em hexadecimal: 0xff
print(int(hex(0xff),16)) # converte o número hexadecimal 0xff em inteiro
````
