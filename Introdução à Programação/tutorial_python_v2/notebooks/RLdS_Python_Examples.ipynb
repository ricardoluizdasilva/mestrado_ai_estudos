## `examples/EX-11_Module_SOCKET.py`

```python
#!/usr/bin/python3
#coding: utf-8

# Module SOCKET

# Import module
import os, sys, socket

# Clean the screen
os.system('clear')

HOST = "127.0.0.1"  # Standard loopback interface address (localhost)
PORT = 8085  # Port to listen on (non-privileged ports are > 1023)

with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
    s.bind((HOST, PORT))
    s.listen()
    conn, addr = s.accept()
    with conn:
        print(f"Connected by {addr}")
        while True:
            data = conn.recv(1024)
            if not data:
                break
            conn.sendall(data)
```

## `examples/EX-24_Module_THREADING.py`

```python
#!/usr/bin/python3
#coding: utf-8

# Module threading - example with threading

import os, sys
from time import sleep, perf_counter
from threading import Thread

def replace(filename, substr, new_substr):
    print(f'Processing the file {filename}')
    # get the contents of the file
    with open(filename, 'r') as f:
        content = f.read()

    # replace the substr by new_substr
    content = content.replace(substr, new_substr)

    # write data into the file
    with open(filename, 'w') as f:
        f.write(content)

# Clean the screen
os.system('clear')

def main():
    filenames = [
        'file1.txt',
        'file2.txt',
        'file3.txt',
        'file4.txt',
        'file5.txt',
        'file6.txt',
        'file7.txt',
        'file8.txt',
        'file9.txt',
        'file10.txt',
    ]

        # create threads
    threads = [Thread(target=replace, args=(filename, 'IDS NAME', 'name is'))
            for filename in filenames]

    # start the threads
    for thread in threads:
        thread.start()

    # wait for the threads to complete
    for thread in threads:
        thread.join()

if __name__ == "__main__":
    start_time = perf_counter()

    main()

    end_time = perf_counter()
    print(f'It took {end_time- start_time :0.2f} second(s) to complete.')

```

```python
#!/usr/bin/python3
#coding: utf-8

# Module threading - multiple threads

import os, sys
from time import sleep, perf_counter
from threading import Thread

def task():
    print('Starting a task...')
    sleep(1)
    print('done')

os.system('clear')

start_time = perf_counter()

# create two new threads
t1 = Thread(target=task)
t2 = Thread(target=task)

# start the threads
t1.start()
t2.start()

# wait for the threads to complete
t1.join()
t2.join()

end_time = perf_counter()

print(f'It took {end_time- start_time: 0.2f} second(s) to complete.')
```

## `examples/EX-27_Module_THREADING.py`

```python
#!/usr/bin/python3
#coding: utf-8

# Module threading - parsing arguments to threads

import os, sys
from time import sleep, perf_counter
from threading import Thread

def task(id):
    print(f'Starting the task {id}...')
    sleep(1)
    print('done')


os.system('clear')

start_time = perf_counter()

# create and start 10 threads
threads = []
for n in range(1, 11):
    t = Thread(target=task, args=(n,))
    threads.append(t)
    t.start()

# wait for the threads to complete
for t in threads:
    t.join()

end_time = perf_counter()

print(f'It took {end_time- start_time: 0.2f} second(s) to complete.')
```

## `examples/EX-29_Module_THREADING.py`

```python
#!/usr/bin/python3
#coding: utf-8

# Module threading --> thread_list(), thread append, start, join

import os, sys, logging, threading, time

def thread_function(name):
    logging.info("Thread %s: starting", name)
    time.sleep(2)
    logging.info("Thread %s: finishing", name)

# Clean the screen
os.system('clear')

format = "%(asctime)s: %(message)s"
logging.basicConfig(format=format, level=logging.INFO,datefmt="%H:%M:%S")

threads = list()
for index in range(10):
    logging.info("Main     : create and start thread %d.", index)
    x = threading.Thread(target=thread_function, args=(index,))
    threads.append(x)
    x.start()

for index, thread in enumerate(threads):
    logging.info("Main     : before joining thread %d.", index)
    thread.join()
    logging.info("Main     : thread %d done", index)
```

## `examples/EX-32_Module_THREADING.py`

```python
#!/usr/bin/python3
#coding: utf-8

# Module threading - race condition solution

import os, sys, logging, threading, time, concurrent.futures
from traceback import format_tb

class FakeDatabase:
    def __init__(self):
        self.value = 0
        self._lock = threading.Lock()

    def update(self, name):
        logging.info("Thread %s: starting update", name)
        #logging.debug("Thread %s: about to lock", name)
        with self._lock:
            #logging.debug("Thread %s: has lock", name)
            local_copy = self.value
            local_copy += 1
            time.sleep(0.1)
            self.value = local_copy
            #logging.debug("Thread %s: about to release lock", name)
        #logging.debug("Thread %s: after release", name)
        logging.info("Thread %s: finishing update", name)

if __name__ == "__main__":

    # Clean the screen
    os.system('clear')

    #logging.getLogger().setLevel(logging.DEBUG)
    #logging.basicConfig(filename='debugfile.txt')
        
    format = "%(asctime)s: %(message)s"
    logging.basicConfig(format=format, level=logging.INFO,
                        datefmt="%H:%M:%S")

    database = FakeDatabase()
    logging.info("Testing update. Starting value is %d.", database.value)
    with concurrent.futures.ThreadPoolExecutor(max_workers=2) as executor:
        for index in range(2):
            executor.submit(database.update, index)
    logging.info("Testing update. Ending value is %d.", database.value)
```

```python
#!/usr/bin/python3
#coding: utf-8

# Module threading - Example no threading

import os, sys
from time import sleep, perf_counter
from threading import Thread

def replace(filename, substr, new_substr):
    print(f'Processing the file {filename}')
    # get the contents of the file
    with open(filename, 'r') as f:
        content = f.read()

    # replace the substr by new_substr
    content = content.replace(substr, new_substr)

    # write data into the file
    with open(filename, 'w') as f:
        f.write(content)

# Clean the screen
os.system('clear')

def main():
    filenames = [
        'file1.txt',
        'file2.txt',
        'file3.txt',
        'file4.txt',
        'file5.txt',
        'file6.txt',
        'file7.txt',
        'file8.txt',
        'file9.txt',
        'file10.txt',
    ]

    for filename in filenames:
        replace(filename, 'ID', 'IDS NAME')

if __name__ == "__main__":
    start_time = perf_counter()

    main()

    end_time = perf_counter()
    print(f'It took {end_time- start_time :0.2f} second(s) to complete.')

```
