# Ficheiro

A memória do computador é de acesso randômico e volátil (RAM). Quando o computador é desligado os dados em memória são perdidos. Para manter os dados armazenados e acessíveis posteriormente, após o reínicio do computador, os dados em memória são guardados em disco, e para tal usa-se ficheiros.

O que esses dados representam depende do seu formato, que normalmente é representado por uma extensão. Por exemplo, um ficheiro que possui uma extensão `.pdf` provavelmente está em conformidade com a especificação do Adobe Documents.

Um fichieiro é um conjunto contíguo de bytes usado para guardar dados na memória não-volátil (disco). Estes dados são organizados em um formato específico e podem ser tão simples como um ficheiro de texto ou complicados como um binário (programa executável). Os ficheiros são traduzidos em `binários`, `0 e 1`, para facilitar o processamento pelo computador.

A gestão dos ficheiros é feita, na maioria dos sistemas computacionais modernos, pelo sistema de ficheiros.

Um ficheiro é composto por três partes principais:

- **Header:** Metadados sobre o conteúdo do ficheiro (nome, tamanho, tipo e assim por diante).
- **Dados:** Conteúdo do ficheiro.
- **End of File (EOF):** Caráter especial que indica o fim do ficheiro.

Para aceder um ficheiro em um sistema operativo, um caminho é necessário. O caminho do ficheiro é uma `string` que representa a sua localização. Ele é dividido em três partes principais:

- **Caminho da pasta/diretoria:** A localização da pasta de ficheiros no sistema de ficheiros onde as pastas subsequentes são separadas por uma barra `/` (Linux/Unix) ou barra invertida `\` (Windows), precedido da unidade de disco (i.e. `C:\`).

- **Nome do ficheiro:** Representa o nome do ficheiro.
- **Extensão:** O final do caminho do ficheiro prefixado com um ponto `.` usado para indicar o seu tipo, na maioria dos casos associado a uma aplicação específica.

## Operações em ficheiros

O manuseamento básico de ficheiros consiste em três passos:

- `Abrir o ficheiro e assignar o ficheiro à uma variável.`
- `Aceder ao ficheiro:`
    - *Ler do ficheiro (read access)*
    - *Escrever para o ficheiro (write access)*
- `Fechar o ficheiro`


A função **open()** pode ser utilizada para `abrir um ficheiro`, e para tal é necessário informar o caminho onde o ficheiro está localizado, caso este não esteja localizado na mesma diretoria/pasta do programa. 

Para as `demais operações` existem alguns `métodos`, tais como `.write()`, `.read()` ou `.close()`. Existem outros módulos que permitem funções adiconais para tipos de ficheiros especiais, tais como *.csv* ou *.json*, por exemplo.

Adicionalmente, o modo pelo qual o ficheiro é aberto pode ser opcional. Abaixo a **lista** com os `modos` mais importantes:

| Modo | Descrição                                         | Erro                             | Warning                 |
| :--- | :------------------------------------------------ | :------------------------------- | :--------------------------------- |
| r    | Read-only                                         | Write access & file non-existent |  |
| w    | Write-only; new file, if no file present          | Read access                      | Overwrites existing content       |
| a    | Append new content to old content                 | &nbsp;                           |   |
| r+   | Read & write access                               | File non-existent                |                                   |
| w+   | Write & read access; new file, if no file present | &nbsp;                           | Overwrites existing content       |

Quando adicionado um *b* ao modo, o ficheiro será criado cini binário e não texto. Se o modo não for especificado o valor padrão será `"r"`.

**Recomendação:** Especificar *sempre* um modo. Isto simplifica a manutenção do programa. O modo é sempre definido como uma string, i.e. `"r"` ou `'r'`, e não `r`, por exemplo.

### `Abrir ficheiros com with`

Como vimos nos exemplos anteriores, um ficheiro é sempre fechado depois de aberto. Deixar um ficheiro aberto poderá causar erros. 

Para evitar isto utiliza-se um comando que está disponível em *Python*, `with`. Depois de sair do bloco `with`, o ficheiro é fechado automaticamente. Isto permite uma melhor gestão do código, porque permite trabalhar dentro do bloco utilizado para abrir o ficheiro.

Um ficheiro é organizado sequencialmente. E para processar sequências, iterar sobre os seus conteúdos, ou seja linha a linha de um ficheiro, utiliza-se o `for`, como mencionado em outras ocasiões.

Vejamos o **exemplo** abaixo.

```python
# Abrir o ficheiro com open
with open("academia.txt", "r") as file:
    # le o ficheiro linha a linha e imprime o resultado
    for line in file:
        print(line)

# O ficheiro é fechado automaticamente
```


O resultado do exemplo acima foi o esperado? Que tal comparar o resultado da impressão com o conteúdo do ficheiro?

Linhas em branco foram adicionadas entre as linhas existentes no ficheiro durante a impressão. 

Isto ocorre porque ao final de cada linha há uma quebra de linha `\n`. Isto é apenas visível indiretamente, porque o texto continua na próxima linha. Sendo assim, durante a impressão a função `print` adiciona uma linha em branco durante a impressão.

Este comportamento pode ser corrigido de várias maneiras. Uma delas é defini o parâmetro `end=""` na função `print()`.

A alternativa é utilizar o método `strip()` para remover os espaços, tabulações e quebra de linhas no início e fim da string. Este método é frequentemente usado na leitura de formulários como forma de prevenir que se altere a entrada de dados. E através de um argumento é possível remover um caráter específico.

Outras alternativas são `.lstrip()` ou `.rstrip()`, utilizados para remover um caráter apenas a esquerda ou direita da string, respetivamente.


```python
# Abrir o ficheiro com open
with open("academia.txt", "r") as file:
    # le o ficheiro linha a linha e imprime o resultado
    for line in file:
        print(line, end="")
        #print(line.strip())

# O ficheiro é fechado automaticamente
```


O que acontece se executar o loop `for` duas vezes?


```python
# Abrir o ficheiro com open
with open("academia.txt", "r") as file:
    # le o ficheiro linha a linha e imprime o resultado
    print("Primeira execução!")
    for line in file:
        print(line, end="")
        #print(line.strip())
        
    print("\nSegunda execução!")
    for line in file:
        # print(line, end="")
        print(line.strip())
```


Ao ler um ficheiro, `cursor de leitura` ou `ponteiro de leitura` é movido caráter a caráter, percorrendo assim todo o ficheiro. Quando o ponteiro chega ao *fim do ficheiro a leitura não continua* porque não é feito um reposicionamento do do ponteiro.

O método `.seek(offset, from_what)` pode ser utilizado para reposicionar o `cursor de leitura`.

- `Offset:` Número de posições para mover o ponteiro
- `from_what:` Define a posição de referência no ficheiro

A posição de referência aceita três valores:
- `0:` posiciona no início do ficheiro 
- `1:` posiciona na posição atual do ficheiro 
- `2:` posiciona no fim do ficheiro


Já o método `.tell()` indica em que posição o cursor de leitura se encontra.


```python
with open("academia.txt", "r") as file:
    # le o ficheiro linha a linha e imprime o resultado
    print("Primeira execução!")
    for line in file:
        print(line, end="")
        #print(line.strip())
    print("Posição: ", file.tell())
    pos_file = file.seek(15,0)
    print(pos_file)
    
    print("\nSegunda execução!")
    for line in file:
        # print(line, end="")
        print(line.strip())
```

### `Escrever em um ficheiro`

Os métodos `.write()` e `.writeline()` são utilizdos para escrever algo nos ficheiros. A **célula abaixo** exemplifica cada um dos métodos.

```python
with open("ficheiro.txt", "w") as file:
    file.write("Olá futuros programadores de Python!\n")
    file.write("Meu nome é ")
    file.write("Darth Vader!\n")
    file.writelines(["Sou o pai da Lea!", " ", "E do Luke!"])
    file.writelines(["\nEstou a caminho do vosso planeta!", "\nAté já!"])
```


E a **próxima célula**, uma vez mais, demonstra a `leitura de um ficheiro`. Neste caso a leitura será feita no ficheiro criado.


```python
file = open('ficheiro.txt', "r")
data = file.read(40)
print(data)
data = file.read(24)
print(data)
for line in file:
    print(line.strip())
file.close()
```

No **exemplo abaixo** algumas `operações de leitura e posicionamento` no ficheiro.

```python
# Posiciono o índice na posição 26
with open("ficheiro.txt", "r") as file:
    file.seek(26)
    print(file.tell())
    print(file.readline())

# Abre o ficheiro em modo binário
with open("ficheiro.txt", "rb") as file:
    # Posiciona o ponteiro na décima posição da esquerda até o fim
    # Define o ponteiro de referência no fim do ficheiro
    file.seek(-18,2)
    print(file.tell())
    print(file.readline())
    # Converte de binário para string
    print(file.readline().decode('utf-8'))


# Abre o ficheiro em modo binário
with open("ficheiro.txt", "rb") as file:
    # Posiciona o ponteiro na décima posição da esquerda até o fim
    # Define o ponteiro de referência no fim do ficheiro
    print("\n")
    file.seek(-18,2)
    print(file.tell())
    # Converter de binário para string
    print(file.readline().decode('utf-8'))
    print(file.readline())
    
```
