
# Boas práticas de programação

A **organização do código** torna tudo mais `fácil e legível`!




## Sintaxe
Python foi desenvolvido como uma linguagem de ensino. Contudo, muitos iniciantes e especialistas das mais diversas áreas adotaram esta linguagem de programação devido a sua facilidade de uso e simplicidade e clareza de sua sintaxe.

A sintaxe da linguagem de programação Python é o `conjunto de regras` que define como um programa Python será escrito e interpretado.

A `clareza do código` deve ser válida tanto para o sistema de execução quanto para o ser humano.



## Shebang
Uma linha `shebang` consiste na sequência de caracteres, sendo os mais conhecidos o numério e o ponto de  exclamação `#!`, e são os primeiros caracteres de um ficheiro de texto, especificamente em um código fonte em uma linguagem interpretada ou scripting.

O `shebang` é muito útil quando utilizado em sistemas Linux e Unix, permitindo gerir várias instalações de Python, ou seja, utilizar aquela que é necessário ao projeto em desenvolvimento.

Em sistemas Windows a sua utilziação não é necessária.



## Identificadores
Os identificadores são nomes dados às entidades como variáveis, funções, classes, etc, e ajudam a diferenciar uma entidade da outra.



## Regras
Os identificadores podem ser letras, dígitos ou underline `_`.

Há uma série de regras para dar nome aos identificadores em Python. O nome pode começar com uma letra ou um sublinhado _ e pode ser complementado por qualquer número de letras, dígitos e sublinhados. É possível utilizar número no nome de um identificador, desde que não esteja no início do nome.

O nome de um identificador não pode conter caracteres especiais ou espaços. No decorrer do curso iremos ver vários casos, como por exemplo dar nome às variáveis.



## Identação
A identação em Python é importante porque indica blocos de código. A não utilização poderá causar erros.

Algumas regras de identação:
- Usar dois pontos `:` para iniciar um bloco e pressione `[enter]`;
- Todas as linhas em bloco devem usar a mesma identação, seja comm espaços ou `[tab]`;
- Recomenda-se quatro espaços como identação para tornar o código mais legível. Não misture `espaço` e `[tab]` no mesmo bloco.



## Comentários
É possível incluir comentários num programa Python. Isto ajuda a entender o que foi feito quando o programa é lido por outro programdor.

Os comentários começam por `#`, como no exemplo abaixo.

```python
# Este é um comentário
print("Bem vindo ao curso básico de Python")
```

Também suporta docstrings, que são comentários em várias linhas.

```python
"""
Este é um exemplo de comentário
que abrange várias
linhas do programa!
"""
print("Procure sempre praticar boas práticas e documentar o seu código")
```


## Statements
As instruções em Python normalmente terminam com uma nova linha.
Em Python, entretanto, permite o uso do caracter de continuação de linha `\` para indicar que a linha deve continuar. Por exemplo:

```python
total = 5 + \
    5 + \
    8 - \
    3
print(total) # 15
```

Já o ponto é vírgula `;` permite várias instruções em uma única linha, porque nenhuma instrução inicia um novo bloco de código, conforme exemplificado abaixo:

```python
x,y = 0, 1; z = x * y + x; print(z)
# 0 * 1 + 1 = 1
```

É possível, através de **duas linhas de código**, obter as `palavras-chave` através do interpretador Python:


````python
import keyword
print(keyword.kwlist)
````
