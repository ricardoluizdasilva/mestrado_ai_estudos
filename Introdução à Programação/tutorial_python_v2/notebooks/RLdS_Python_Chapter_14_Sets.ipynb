# Sets

Sets são coleções não-ordenadas de itens únicos não indexados. Embora um set possa ser modificado, os elementos contidos dentro do `set` devem ser `imutáveis`. Os Sets são definidos por valores separados por vírgula dentro de chaves `{}`.

O `set`, ao contrário de `list`, possui um `mecanismo otimizado` para verificar se um elemento específico está contido nele.

Podem ser utilizados nas operações matemáticas de conjuntos como união, interseção, diferença, simétrica, etc.

A **tabela** a seguir apresenta os `métodos` mais importantes dos sets.




## Operações básicas

### `Criar um set`


```python
n_ips = {"192.168.0.1", "192.168.0.112", "192.168.0.233", "192.168.0.1"}
print(type(n_ips))
print(n_ips)
```

Também pode ser criado através do `método set`.

```python
nsets = set([x for x in range(10)])
print(nsets)
```

E, a utilização do `for` permite **iterar sobre os elementos de um set**.

```python
for itens in nsets:
    print(itens, end = " ")
```

E mais uma vez o método `len()` é utilizado para determinar o `número de elementos` em um set.

```python
print(len(nsets))
```

Os `operadores in e not in` também podem ser utilizados com `sets`.

```python
print(1 in nsets)
print(4 in nsets)
print(10 not in nsets)
print(3 not in nsets)
```

### `Método add()`

Permite adicionar um elmento ao `set`.

```python
nsets.add(11)
print(nsets)
```

### `Método discard()`

Permite `remover um` elemento do `set`.


```python
nsets.discard(0)
print(nsets)
```

### `Método clear()`

Permite `remover todos` os elementos do `set`, retornando um `set() vazio`.

```python
n_ips.clear()
print(n_ips)
```

O `set` também pode ser utilizado para `eliminar letras repetidas` de uma *string*.

```python
print(set("O rato roeu a roupa do Rei de Roma"))
```

### `Intersecção`

Para a *intersecção* utiliza-se o `operador &` ou o `método intersection()`.


```python
x = {1, 2, 3, 4 ,5, 7, 9}
y = {0, 2, 4, 6, 8, 10}

print(x & y)
print(x.intersection(y))
```

### `Diferença`

Para a *diferença* utiliza-se o `oprerador -` ou o `método difference()`.

```python
# Relativo a x
print(x - y)
print(x.difference(y))

# Relativo a y
print(y - x)
print(y.difference(x))
```

### `União`

Para *união* utilizar-se o `operador |` ou o `método union()`.


```python
print(x | y)
print(x.union(y))
z = x.union(y)
```

### `Método remove()`

Este método `remove` o elemento especificado do set.

```python
z.remove(10)
print(z)
```

### `Método pop()`

Este método `remove` o primeiro elemento do set.

```python
z.pop()
print(z)
```
