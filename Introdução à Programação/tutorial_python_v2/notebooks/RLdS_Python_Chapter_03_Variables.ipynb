# Variáveis

**O uso de variáveis em programação é essencial!**

As variáveis existem em (quase) todas as linguagens de programação. No entanto, o termo tem um significado diferente na progamação em comparação a matemática.

Na programação, uma variável é um nome ao qual um valor pode ser atribuído. Do ponto de vista técnico, numa visão bastante simplista, uma variável é o espaço em memória que tem um determinado valor num determinado momento no tempo.

Em Python as variáveis armazenam dados e são definidas pela sua atribuição. Não têm um tipo fixo, mas o tipo de conteúdo.


## Usando variáveis

É possível atribuir (ou escrever ou colocar) um valor à variável, bem como ler o valor atualmente armazenado. É importante ressaltar que após a leitura do valor da variável, este continua disponível e inalterado.

O valor de uma variável pode ser definido utilizando o operador de atribuição `=`. Por exemplo, `a = 10` significa que a variável de nome `a` recebe o valor `10`, ou escreve o `valor 10 na variável a`. Se no decorrrer do tempo for atribuído um `novo valor` a variável `a`, este assumirá o novo valor. Por exemplo, `a = 3` significa que foi colocado o `valor de 3 à variável a`.

No **exemplo abaixo**, são atribuídos às variáveis `a` e `b` os valores `10` e `5`, respetivamente. A variável `c` é a soma das duas variáveis.



````python
a = 10
b = 5
c = a + b
c
````

`Nota:` Em matemática, `a = 10` é uma afirmação, que pode ser verdadeira ou falsa. Em programação, `a = 10` é uma instrução, que descreve o `valor 10 atribuído à variável a`.


### `Várias atribuições em sequência`
O valor de uma variável, como já foi dito, pode ser alterado por atribuição, se as atribuições ocorrerem uma após a outra. No **exemplo abaixo**, há uma variável de nome `a`. Primeiro, o valor `5` é atribuído, depois o valor `8` é atribuído. Esta segunda atribuição elimina o valor 5 e, a memória é apenas ocupada com o valor 8.


````python
a = 5
a = 8
a
````

### `Atribuindo múltiplos valores a múltiplas variáveis`

Como vimos anteriormente, no **primeiro exemplo**, é possível ter mais do que uma variável num programa Python, atribuindo às variáveis valores, inclusive por referência a outras variáveis.


````python
a = 10
b = 5
c = a * b
c
b
````

### `Variável com atribuição em ambos os lados`
As atribuições podem parecer estranhas (ou mesmo erradas) através dos olhos da matemática. É o caso, por **exemplo**, quando uma variável está em ambos os lados da atribuição. No exemplo que se segue, a variável `a` recebe o valor de `10` e posteriomente é incrementada em `18` valores. Para ser mais preciso, primeiro o valor da variável `a` é lido, e o resultado final (valor anterior mais a nova atribuição) é escrito de volta na memória.


````python
a = 10
a = a + 18
a
````

### `Sequenciação de atribuições`
Por vezes poderá acontecer de ter várias operações aplicadas a uma variável. Estes também podem ser sequenciados conforme vimos acima e iremos ver abaixo.

Esta abordagem permite tornar o programa mais legível. Por exemplo, a atribuição `a = 5 * 3 * 8 + (2 - 8 * 3)` pode ser **escrita da seguinte forma** (quando sequenciado, deve-se ter cuidado para não violar nenhuma regra de síntaxe):


````python
a = 5
a = a * 3
a = a * 8
a = a + (2 - 8 * 3)
a
````

### `Apenas uma variável no lado esquerdo de uma atribuição`
No lado direito de uma atribuição de valor à uma variável, pode haver expressões complexas, que por si só, contêm variáveis `(Exemplo, a = b ** 5 + c ** 2)`. Do **lado esquerdo**, por outro lado, `só pode haver uma única variável`. Caso contrário, ocorrerá um erro.

````python
b = 3
c = 4

a ** 2 = b ** 2 + c ** 2
a
````

Provavelmente, você tenha reconhecido o teorema de `Pitágoas` na célula acima. Mas, como obter o `comprimento de a se b e c são dados`? O cálculo pode ser feito em dois passos, conforme **exemplificado** abaixo.


````python
b = 3
c = 4

a = b ** 2 + c ** 2
a = a ** 0.5
a
````

## Tipos de variáveis
Em muitas linguagens de programação, o tipo de variável é estático e atribuído na sua definição. Isto significa que uma variável é inicialmente declarada como um tipo de dados específico, e qualquer valor que lhe seja atribuído durante a sua existência deve ter sempre esse tipo.

Em Python a variável não tem um tipo fixo, mas o tipo de conteúdo. Ou seja, pode ser atribuído um valor que define seu tipo e, em determinado momento, o seu valor é alterado para um tipo de dados diferente.

Por exemplo, `a = 10` significa que a variável `a tem o valor de 10`, `tipo numérico (inteiro)`. Posteriormente, o seu valor é alterado para `"casa" (a = "casa")`. Consequentemente, o seu tipo de dados foi alterado para `alfanumérico (string)`.

Estes são os tipos de variáveis mais utilizados:




**Definições** (não retrata a síntaxe da função ou método):
- **integer:** it is a number, the whole number, positive or negative, without any decimals or unlimited lenght.
- **float:** it is a function, that returns a number, positive or negative, containing one or more decimals.
- **string:** it is an array of bytes representing unicode characters.
- **boolean:** is is logical and represents one of two values, True or False
- **file:** it is method that opens a file and assign it to a variable.
- **list:** it is an ordered sequence of objects; it is just like a dymamic sized array.
- **dicitionary:** it is a colletcion or associative arrray which is used to store vaous types of data. It contains pairs of objects (key and value).
- **tuple:** it is similar to a list in terms of indexing, nested objects, and repetition. But a tuple is imutable.
- **set:** it is unordered collection data type that is iterable, immutable, unchangeable, unindexed, and has no duplicate elements.
- **frozenset:** it returns an imutable frozenset object initialized with elements from the given iterable.
- **range:** a range of numbers (i.e. - from 0 to 9)
- **NoneType:** it is an object that indicates no value.

## Mais detalhes sobre variáveis

Em Python, as variáveis não têm de ser declaradas primeiro, como em algumas outras linguagens de programação. As variáveis podem ser criadas quando forem ser utilizadas.

Uma **variável** pode estar no `lado direito e esquerdo do operador de atribuição` ao mesmo tempo.


````python
a = 10
a = a + 1
a
````

No **exemplo acima**, o valor da variável `a é incrementado em 1`.

Em Python, existe outra alternativa, que é o valor por atribuição, combinando dois operadores: `a += 1`

Do ponto de vista matemático isto parece muito estranho, contudo em Python (e outras linguagens de programa) isto significa simplesmente que o valor armazenado na variável a é aumentado em 1.

Esta abordagem também se aplia a outros operadores, como se pode observar na **próxima célula**.


````python
a = 100
a *= 2
a -= 100
a
````

## Variáveis globais
Uma variável quando declarada tem a sua definição e valor definidos na função onde se encontra, ou seja, uma outra função não terá acesso variável. Iremos estudar funções em mais detalhes durante o curso.

Em Python, uma variável declarada como global pode ser acedida dentro ou fora da função onde foi declarada.

Exemplo de como definir uma variável global:

```python
x = 5

def multiplica():
    global x
    x = x * 3
    print(x)

multiplica() # 15
print(x - 3) # 12
```

Utilize a **célula abaixo** para `executar as instruções do exemplo acima`.

Abaixo um **outro exemplo**, onde as variáveis estão definidas dentro e fora da função.

````python
nome = "Obi-Wan Kenobi"


def func_nome(nome):
    print("O seu nome é", nome)

func_nome("Dart Vader")
          
print("O seu nome é", nome)
````


No **próximo exemplo** a variável `nome` é declarada como *global*. Vejamos como o programa se comportará.


````python
nome = "Obi-Wan Kenobi"


def func_nome():
    global nome
    nome = "Darth Vader"
    print("O seu nome é", nome)

func_nome()
          
print("O seu nome é", nome)
````
