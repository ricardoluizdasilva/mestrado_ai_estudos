# O que é Python?



É uma `linguagem de programação`. Foi criada em `1991` por `Guido Van Rossumen`, com o objetivo de ter uma `linguagem simples, de fácil compreensão e aprendizado`.

A linguagem de **programação Python** permite:
- **Legibilidade:** não há uso excessivo de marcações (ponto, ponto e vírgula) e marcadores (chaves, colchetes, parenteses).
- **Portabilidade:** multiplataforma
- **Software livre:** há muita colaboração entre muitas entidades – mantida pelo Python Software Foundation (PSF)

O `Python é uma linguagem de programação de alto nível`. Isto significa que não temos de dar instruções demasiado concretas como por exemplo movimentação de objetos de memória para imprimir uma string. Em Python, basta escrever `print` para que possamos fazer a impressão para o terminal.

À semelhança de todas as outras linguagens de programação, serve para automatizar tarefas e economizar tempo. 

É extremamente utilizada em Inteligência Artificial, Data Science, Big Data, Visão por Computador, Automatização.

Diferencia-se das restantes por ser de alto nível e extremamente intuitiva, para além de ser multi-plataforma. 

É utilizada pela NASA, Intel, Cisco, HP, Seagate, etc.



## História da linguagem Python

**A linguagem de programação Python** foi concebida no `final da década de 1980` por `Guido van Rossum`, enquanto trabalhava no `Centrum Wiskunde & Informatica (CWI)`, na Holanda. O Python surgiu como `sucessor da linguagem de programação ABC`, com o intuito de oferecer uma alternativa mais flexível e poderosa para o desenvolvimento de software.

Van Rossum desenvolveu o Python inspirado pela `necessidade de uma linguagem de mais alto nível no projecto do sistema operativo Amoeba`. O desenvolvimento de `utilitários para administração de sistemas em C` mostrava-se `demasiado moroso`, e a utilização do `Bourne Shell` apresentava `diversas limitações`. Assim, o `Python foi criado para preencher esta lacuna`, posicionando-se entre a `eficiência da linguagem C` e a `simplicidade de utilização dos scripts de shell`.

A primeira versão pública, `Python 0.9.0`, foi disponibilizada em fevereiro de 1991. Já nesta fase inicial, Python incluía funcionalidades como exceções, módulos e tipos de dados como listas e dicionários. O lançamento oficial da versão 1.0 ocorreu em 1994, introduzindo novas capacidades, como as funções lambda, e os métodos map(), filter() e reduce().

No ano 2000, foi lançada a versão `Python 2.0`, trazendo melhorias significativas, como a introdução das listas por compreensão (list comprehensions) e mecanismos de recolha automática de lixo (garbage collection), facilitando a gestão da memória. Apesar das melhorias, o `Python 2` permaneceu com algumas limitações de design, que viriam a ser resolvidas na versão seguinte.

O `Python 3.0` foi disponibilizado `em dezembro de 2008`, marcando uma `grande reformulação da linguagem`. Esta versão trouxe diversas melhorias, incluindo uma melhor gestão de caracteres Unicode, um novo modelo de manipulação de exceções e a unificação da divisão inteira e de ponto flutuante. No entanto, devido às mudanças fundamentais, `Python 3 não era retrocompatível com Python 2`, o que levou a um longo período de coexistência entre ambas as versões.

Desde a sua criação, `o Python sempre foi um projeto de código aberto (open source)`, permitindo que a comunidade global contribuísse para o seu desenvolvimento e expansão. A simplicidade da sintaxe, a legibilidade do código e a sua versatilidade tornaram Python uma das linguagens de programação mais populares do mundo, sendo amplamente utilizada em diversas áreas, como desenvolvimento web, análise de dados, inteligência artificial e automação de sistemas.

Atualmente, o `Python` continua a `evoluir`, com `novas versões e funcionalidades` que o mantêm `relevante e inovador`. A sua popularidade crescente e o `forte apoio da comunidade` garantem que a linguagem continuará a desempenhar um papel essencial no panorama tecnológico global.



## Linguagem de Programação e Scripting

Um **programa é executado** (isto é, a fonte é compilada pela primeira vez, e o resultado dessa compilação é esperado). Um programa em geral é uma sequência de instruções escritas para que um computador possa executar determinada tarefa.

O **Scripting é interpretado** e seu código é escrito numa linguagem de script. Uma linguagem de script não é nada mais do que um tipo de linguagem de programação em que podemos escrever código para controlar outra aplicação de software.


## O que é um Programa Python?

**O que é um programa Python?** De uma forma mais geral, o que é um programa de computador? Todos os programas Python consistem numa série de declarações ou instruções. E estas declarações são executadas uma a uma. `(Esta ordem de execução pode ser alterada, por exemplo, por condições, que serão discutidas posteriormente)`.

Um `exemplo de tal afirmação` é a multiplicação de dois números. A multiplicação é representada em Python pelo símbolo *.
Se apenas uma declaração for inserida, o programa Python pode ser usado como uma simples calculadora. Se alterar os valores ou o operador (isto é, o * ou o +), pode fazer cálculos simples.

O **Python é linguagem de programação de alto nível**, dinâmica, interpretada, funcional, imperativa, procedural, modular, orientada a objetos e multiplataforma.

Possui uma tipagem dinâmica e um sistema de gestão de memória automático, sendo capaz de suportar múltiplos paradigmas de programação (https://pt.wikipedia.org/wiki/Paradigma_de_programa%C3%A7%C3%A3o).

É linguagem de programação, bem como linguagem de script.



- Quando um script *.py é executado via Python, é compilado para ByteCode e armazenado com extensão *.pyc.
- Depois, é interpretado dentro da Máquina Virtual Python (PVM).
- Podemos dizer que é uma linguagem compilada e interpretada!

A `linguagem de programação de baixo nível` é uma linguagem de programação que segue as características da arquitetura do computador. Assim, utiliza somente instruções que serão executadas pelo processador, em contrapartida as linguagens de alto nível que utilizam de instruções abstratas. Nesse sentido, `as linguagens de baixo nível estão diretamente relacionadas com a arquitetura do computador`. Como exemplos, temos Binário e Assembly

As linguagens de baixo nível são divididas em duas categorias: `primeira geração e segunda geração.`

A `linguagem de primeira geração`, ou 1GL, é o código de máquina. É a única linguagem que um microprocessador pode executar nativamente. O código de máquina pode ser escrito diretamente por um editor hexadecimal. Atualmente é raro que um programador trabalhe com o código de máquina diretamente, preferindo o uso da linguagem Assembly.

A `linguagem de segunda geração`, ou 2GL, é a linguagem Assembly. É considerada de baixo nível pois embora não seja uma linguagem nativa do microprocessador, ainda assim utiliza apenas de instruções reais do processador. Desta forma a linguagem Assembly não contém nenhum tipo de instrução abstrata, onde cada instrução é convertida diretamente para a instrução equivalente em código de máquina pelo montador.

A `linguagem de programação de médio nível`, que possui símbolos que podem ser convertidos diretamente para código de máquina (goto, expressões matemáticas, atribuição de variáveis), mas também símbolos complexos que são convertidos por um compilador. Exemplo: C, C++

A `Linguagem de programação de alto nível, é uma linguagem com um nível de abstração relativamente elevado, longe do código de máquina e mais próximo à linguagem humana. Desse modo, as linguagens de alto nível não estão diretamente relacionadas à arquitetura do computador. O programador de uma linguagem de alto nível não precisa conhecer características do processador, como instruções e registradores. Essas características são abstraídas na linguagem de alto nível.

A Linguagem de alto nível é inteligível pelo ser humano e não executável diretamente pela máquina, no nível da especificação de algoritmos, como Pascal, Fortran, ALGOL,Java e SQL.

Por se tratar de uma `classificação subjetiva`, isto é, sem limites bem definidos, não é possível afirmar que `determinada linguagem pode ser mais humana que outra`. Apesar disso, por questão de praticabilidade e objetividade, a classificação geralmente se limita em `linguagem de alto nível` e `linguagem de baixo nível`.



## Filosofia do Python

**Zen of Python** é uma coleção de **19 princípios orientadores** para a escrita de programas de computador que influenciam o desenho `(design)` da linguagem de programação `Python`.

Quando concebida, um dos seus objetivos era de `ser uma linguagem divertida de usar`. Isto reflete-se no seu nome - uma homenagem ao grupo de comédia britânica Monty Python - e em abordagens ocasionalmente lúdicas a tutoriais e materiais de referência, tais como exemplos que se referem a `spam and eggs` (uma referência a um esboço de Monty Python) em vez do padrão `foo and bar`.

**Zen of Python em inglês:**
1. Beautiful is better than ugly.
2. Explicit is better than implicit.
3. Simple is better than complex.
4. Complex is better than complicated.
5. Flat is better than nested.
6. Sparse is better than dense.
7. Readability counts.
8. Special cases aren't special enough to break the rules.
9. Although practicality beats purity.
10. Errors should never pass silently.
11. Unless explicitly silenced.
12. In the face of ambiguity, refuse the temptation to guess.
13. There should be one– and preferably only one –obvious way to do it.
14. Although that way may not be obvious at first unless you're Dutch.
15. Now is better than never.
16. Although never is often better than right now.
17. If the implementation is hard to explain, it's a bad idea.
18. If the implementation is easy to explain, it may be a good idea.
19. Namespaces are one honking great idea – let's do more of those!

**Zen of Python em Português**
1. O belo é melhor do que o feio.
2. O explícito é melhor do que o implícito.
3. O simples é melhor do que o complexo.
4. O complexo é melhor do que o complicado.
5. O plano é melhor do que o aninhado.
6. O espaçado é melhor do que o denso.
7. A legibilidade conta.
8. Casos especiais não são especiais o suficiente para quebrar as regras.
9. Embora a praticidade supere a pureza.
10. Os erros nunca devem passar silenciosamente.
11. A menos que sejam silenciados explicitamente.
12. Diante da ambiguidade, recusa a tentação de adivinhar.
13. Deve haver uma — e preferencialmente apenas uma — maneira óbvia de o fazer.
14. Embora essa maneira possa não ser óbvia à primeira vista, a menos que sejas holandês.
15. Agora é melhor do que nunca.
16. Embora o nunca seja muitas vezes melhor do que já.
17. Se a implementação for difícil de explicar, é uma má ideia.
18. Se a implementação for fácil de explicar, pode ser uma boa ideia.
19. Namespaces são uma ideia excelente — vamos usar mais disso!
