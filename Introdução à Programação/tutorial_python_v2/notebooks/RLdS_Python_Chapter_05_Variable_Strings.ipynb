# Strings


## Definição

É um tipo de dados que representa uma `sequência de carateres`, que podem ser letras, digitos, símbolos ($, !, #, @), etc.

Um **caráter** é simplesmente um `símbolo`. Por exemplo, o alfabeto do idioma português possui 26 carateres.

Os computadores lidam fundamentalmente com números `binários`. No ecrã estaremos a ver o caráter, mas internamente eles são armazenados e manipulados como uma combinação de `0 e 1`

A conversão de carateres em números é chamada de `codificação` e o processo inverso é a `decodificação`. `ASCII` e `UNICODE` são algumas das codificações mais usadas. Em `Python`, uma `string` é uma sequência de `carateres Unicode`, que foi introduzido para incluir todos os carateres de todos os idiomais, uniformizando a codificação.





A função `ord()` permite obter o valor ASCII de um determinado caráter.


````python
print(ord('A'))
print(ord('<'))
print(ord('∑'))
print(ord('火'))
````

O método `chr()` faz o `inverso` de `ord()`, ou seja, a partir do `valor númerico` retorna uma `string` representando este valor - o caracter.


````python
print(chr(65))
print(chr(60))
print(chr(8721))
print(chr(28779))
print(chr(960))
````

### `Índices`

Em Python, como em muitas outras linguagens de programação, `strings` são consideradas `arrays de bytes` que representam caracteres unicode. Em Python não há um tipo de dados caráter, portanto um único caráter é uma string de `tamanho 1`.

A função `len()` - buil-in funcition - retorna o tamanho (comprimento) de um objeto. No caso das strings, retorna o número de carateres.



````python
msg = "PROGRAMADORES"
print(msg, "tem", len(msg), "caractéres")
````

Saber quantos carateres tem uma string permite determinar quais são seus índices.



O índice de uma string (lista, tuple, etc) inicia na posição 0. O *índice reverso* inicia com o `negativo` da última posição mais um, conforme exemplificado acima.

Para aceder a uma posição de uma string basta utilizar a string seguida `[]` com o número que indica o `índice`. A sintaxe é `string[n]`.

```python
a = "Olá"
print(a[1]) # retorna l
```

Isto permite `iterar` sobre uma `string`, ou `slicing` - mais detalhes serão abordados em `listas`.

Se a string tem uma única posição, seu índice será `0` ou `-1`.

```python
msg = "a"
print(msg[0]) # ou -1 --> retorna a
```

````python
msg = "PROGRAMADORES"
print(msg[1], msg[6], msg[-6])
````

### `Métodos`

Existem diversos métodos (funções especiais) que permitem manipular strings.

- `strip():` remove todos os espaços extras no início e fim da string.    
- `lower():` retorna a string em lowercase (minúsculas).
- `upper():` retorna a string em uppercase (maiúsculas).
- `swapcase():` retorna a string com caracteres maiúsculos convertidos em minúsculos e vice-versa.
- `title():` retorna a string no qual a primeira letra de cada palavra é convertida para maiúscula e as demais letras permanecem em minúsculas.
- `replace():` substitui uma string por outra string.
- `split():` separa a string em substrings caso haja um separador.
- `join():` retorna a string que resulta da concatenação dos objetos em um iterável separados por um delimitador.
- `count():` retorna o número de ocorrências não sobrepostas de uma substring.
- `startswith():` retorna `True` se uma string inicia com um prefixo especificado.
- `endswith():` retorna `True` se uma determinada string termina com um sufixo especificado.
- `find():` pode ser utilizado para ver se uma string contém uma substring informada e retorna o menor índice na string onde esta substring é encontrada.
- `isalnum():` retorna `True` se a string for `não-vazia` e todos os seus carateres forem `alfanuméricos` (ou uma letra ou um número).
- `isalpha():` retrona `True` se a string for `não-vazia` e todos os seus carateres forem `alfabéticos`.
- `isdigit():` retorna `True` se a string for `não-vazia` e contém apenas `dígitos`.
- `isidentifier():` retorna `True` se uma determina string é um identificador válido de acordo com a definição da linguagem Python.
- `iskeyword():` testa se uma determina string corresponde à uma `palavra-chave` Python. A utilização deste método implica na importação do método keyword.
- `isprintable():` determina se uma string consiste inteiramente de carateres que podem ser impressos. Retorna `True` se a string for vazia ou se todos os carateres alfabéticos forem passíveis de impressão. Os carateres não-alfabéticos são ignorados.
- `isspace():` determina se uma string consiste de carateres de `espaço em branco`. Retornará `True` se a string for não-vazia e todos os carateres forem de espaço em branco.
- `isupper():` verifica se todos os carateres de uma string são maiúsculos.
- `islower():` verifica se todos os carateres de uma string são minúsculos.
- `center():` retorna uma string centralizada em um determinado comprimento especificado via argumento. Por definição padrão, o preenchimento consiste no caráter de `espaço` ASCII.
- `expandtabs():` substitui cada caráter tab `(\t)` com espaços.
- `ljust():` retorna uma string justificada à esquerda em um campo de comprimento especificado via argumento. O preenchimento padrão consiste no caráter de `espaço` ASCII.
- `rjust():` retorna uma string justificada à direita em um campo de comprimento especificado via argumento. Por definição, o preenchimento padrão consiste no caráter `espaço` ASCII.
- `zfill():` retorna uma cópia da string com preenchimento à esquerda com caráteres '0' e o comprimento é especificado como argumento.
- `format():` permite construir strings de uma forma flexível. *Iremos abordar mais detalhes no capítulo específico de formatação de saída de dados.

Todos os métodos *`.issomething()`* retornam valores boolean (`True` ou `False`).

### `Programa para imprimir no ecrã a tabela ASCII completa (0 a 172)`



````python

control_names = [
    "NUL", "SOH", "STX", "ETX", "EOT", "ENQ", "ACK", "BEL",
    "BS", "HT", "LF", "VT", "FF", "CR", "SO", "SI",
    "DLE", "DC1", "DC2", "DC3", "DC4", "NAK", "SYN", "ETB",
    "CAN", "EM", "SUB", "ESC", "FS", "GS", "RS", "US"
]

def ascii_char(n):
    if n < 32:
        return control_names[n]
    if n == 32:
        return "SP"
    if n == 127:
        return "DEL"
    return chr(n)

header = f"{'Dec':<4} {'Hx':<3} {'Oct':<3} {'Char':<5}"
print(" | ".join([header] * 4))
print("-" * 107)

for row in range(32):
    columns = []

    for group in range(4):
        n = row + group * 32
        columns.append(
            f"{n:<4} {n:02X}  {n:03o}  {ascii_char(n):<5}"
        )

    print(" | ".join(columns))
````
