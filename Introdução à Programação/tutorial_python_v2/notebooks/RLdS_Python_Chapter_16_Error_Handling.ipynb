# Tratamento de exceções e erros

O que se pretende com este capítulo *não é tratar exceções e erros de sintaxe* `(parsing errors)`.

O objetivo é **lidar com erros e exceções** que possam acontecer `durante a execução do programa/script`.

Uma *exceção bastante comum é a divisão de um número por 0*. Quando isto acontece o interpretador Python cria um `objeto de exceção`. Vejamos o **seguinte exemplo**:


```python
x = 3
y = 0
print(x / y)
```

Uma `divisão por 0` gera uma exceção **ZeroDivisionError**, conforme observado no exemplo acima. Como esta exceção não foi tratada, o interpretador Python retornou um **traceback com mais detalhes sobre o erro**.

Uma outra exceção bastante comum é quando o program tenta `abrir um ficheiro que não existe`. No **próximo exemplo** o interpretador Python irá gerar uma exceção **FileNotFoundError**.

> **Nota:** a manipulação de ficheiros será abordada no próximo capítulo deste curso.


```python
nfile = open('log.txt', 'r')
```

Existem muitas outras operações ilegais que poderão gerar exceções. A **tabela** a seguir ilustra algumas das `exceções` mais comuns.




## Lidar com exceções

Para lidar com as exceções utilizamos as palavras-chave `try`, `except` e `finally`.

- **try:** testa um bloco de código
- **except:** lida com a exceção
- **finally:** executa o código independentemente do resultado do *try* e do *except*

O **exemplo** a seguir permite ter a primeira noção de como `lidar com exceções`.

```python
x = int(input("Digite um número: "))
y = int(input("Digite outro número: "))

try:
    z = x / y
    print(f"A divisão de {x} por {y} é: {z}!")
except:
    print(f"Não é possível dividir {x} por {y}!")
    
```

E a mesma abordagem poderia ser utilizada para `abrir um ficheiro`, por **exemplo**.


```python
try:
    nfile = open('log.txt', 'r')
    print("Foi possível abrir o ficheiro!")
except:
    print("Não foi possível abrir o ficheiro")
    
```

No *exemplo acima* foi feito o `tratamento da exceção de forma genérica`, utilizando apenas o `try` e `except` *(sem indicar uma exceção)*.

No **próximo exemplo** a abordagem é diferente, pois são tratados `diferentes tipos de exceções`.

```python
try:
    x = int(input("Digite um número: "))
    y = int(input("Digite outro número: "))
    z = x / y
    print(f"A divisão de {x} por {y} é: {z}!")
except ZeroDivisionError:
    print(f"Não foi possível dividir um número por 0!")
except ValueError:
    print(f"Não foi digitado um número!")
except:
    print(f"Não foi possível dividir o número {x} por {y}!")
```

A palvra reservada `finally` vai executar o código independetemente do resultado de *try e except*.

Ao entendermos o `conceito do tratamento de exceções`, conseguimos antecipar eventuais `comportamentos do programa ou até mesmo do utilizador`.

**Exemplo**:


```python
try:
    x = int(input("Digite um número: "))
    y = int(input("Digite outro número: "))
    z = x / y
    print(f"A divisão de {x} por {y} é: {z}!")
except ZeroDivisionError:
    print(f"Não foi possível dividir um número por 0!")
except ValueError:
    print(f"Não foi digitado um número!")
except KeyboardInterrupt:
    print("Interrompido a pedido do utilizador!")
except:
    print(f"Não foi possível dividir o número {x} por {y}!")
```
