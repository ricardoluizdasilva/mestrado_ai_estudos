# Laços (loops)

Em programação, laço (loop) ou estrutura de repetição, é uma instrução para o programa repetir tarefas num mesmo bloco de código.

Em Python existem 2 tipos de loops.





## For

É utilizado para iterar sobre uma sequência de dados como `string`, `range`, `lists`, `sets`, `tupple` e `dictionary`, executando um conjunto de instruções em cada elemento.
 
```Python
for <var> in <iterável>:
    <instruções>
    # break or continue
else:
    <instruções>
```

No **exemplo abaixo** o `for` iterage sobre um `range()`.

````python
for i in range(10):
    print(i, end=' ')
````

### `Range`

A função `range()` permite gerar uma **sequência de números**. No `exemplo acima`, irá gerar números de `0 a 9 (10 números)`. Permite também nos permite definir o começo da sequência, o fim e os passos que serão dados. A sintaxe é `range(start,stop,step)`. Se o **step** não for informado o **padrão é de 1 em 1**, conforme exemplificado anteriormente. No exemplo `abaixo` o **step** `será de 2 em 2`.

````python
for i in range(0, 55, 2):
    print(i, end = ' ')
    
````

No **próximo exemplo** o `for` iterage sobre uma `list`.


````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for name in linux:
    print(name)
    
````


Neste **exemplo**, combinamos o `for` com uma condição.


````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for name in linux:
    print(name)
    if name == "suse":
        break
````

E que tal se o `for` iterar sobre um `dicionário (dict)`. Vejamos **este exemplo:**

````python
dict = {'one': 1, 'two': 2, 'three': 3}
for j in dict:
    print(dict[j])
````

### `Enumerar valores`

O método `enumerate()` adiciona um **contador** a um iterável e retorna o objeto enumerado. Este método recebe dois parâmetros:

- iterable: Uma sequência, um iterator, ou objetos que suportam iteração
- start (opcional): É o número que o método enumerate() iniciará contando, se start for omitido, 0 será o valor padrão.

Na **célula a seguir** um exemplo:

````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for indice,linux in enumerate(linux):
    print(f'{indice} -> {linux}')
````

No `exemplo anterior` a enumeração é a mesma do `indice` de uma lista. No **próximo exemplo** a enumeração o `contador inicia no 1`.

````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for indice,linux in enumerate(linux,1):
    print(f'{indice} -> {linux}')
````

**Sem o enumerate()** a indexação apresentada no 1º exemplo isto poderia ser feito assim:

````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
indice = 0 
for distr in linux:
    print(f'{indice} -> {distr}')
    indice += 1
````

Ou ainda desta **outra maneira**:

````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for indice in range(len(linux)):
    print(f'{indice} -> {linux[indice]}')
````

## While

É utilizada quando não podemos definir a quantidade de iterações que podem ocorrer e quando não há uma sequência a seguir. Repete uma sequência de instruções até que a condição esteja satisfeita (`True`).

```Python
while <expressão>:
    <instruções>
    # break or continue
```



Abaixo um **exemplo** de `while` em que uma variável utiliza um `contador` para incrementar o seu valor.


````python
x = 0
while x < 10:
    print(x, end=" ")
    x += 1 # x = x + 1
````


Um outro **exemplo** é a utilização de um boolean (`True`). Enquanto a condição for verdadeia o `bloco de código` do While é executado.


````python
x = 0
while True:
    print(x, end=" ")
    x += 1 # x = x + 1
    if x == 5:
        break
````

Neste **outro exemplo** `subtraímos um do contador`. E para tal invertemos o `operador da condição` e seu `valor de comparação`.


````python
x = 10
while x > 0:
    print("x = ", x)
    x -= 1 # x = x - 1
````


Vejamos agora um **exemplo** com `continue`.


````python
x = 10
while x > 0:
    x -= 1
    if x == 5:
        continue
    print(x, end=" ")
````


O `while` pode ser utilizado para `validação de dados`. **Por exemplo**, enquanto a condição da entrada de dados não for satisfeita o loop é executado.


````python
x = False
while not x:
    num = input("Digie um número inteiro: ")
    num_sign = "positivo" if bool(num.find("-")) else "negativo"
        
    if num.lstrip("-").isdigit():
        num = int(num)
        print(f"O número digitado {num} é inteiro e {num_sign}" 
              if num != 0 else f"O número digitado é inteiro e {num}")
        x = True
   
````

## For vs While

O `for` é utilizado nas seguintes situações:
- O número de iterações é conhecido
- Pode finalizar antepadamente através do break
- Utiliza um contador


O `while` é utilizado quando:
- O número de iterações é ilimitado
- Pode finalizar antepadmente através do break
- Pode utilizar um contador, porém é necessário inicializar antes do loop e incrementá-lo dentro do bloco de código.
