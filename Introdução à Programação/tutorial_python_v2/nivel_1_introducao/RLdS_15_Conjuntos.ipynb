# Capítulo 15 – Conjuntos e síntese das estruturas de dados

Os conjuntos (`set`) são coleções não ordenadas de elementos únicos, sem índice. Embora um set possa ser modificado (é mutável), os elementos que contém têm de ser `imutáveis`. Os sets escrevem-se com os valores separados por vírgulas dentro de chavetas `{}`.

> **Atenção:** `{}` cria um **dicionário** vazio. Para criar um conjunto vazio usa-se `set()`.

O `set`, ao contrário de `list`, possui um `mecanismo otimizado` para verificar se um elemento específico está contido nele.

Podem ser utilizados nas operações matemáticas de conjuntos como a união, a interseção, a diferença e a diferença simétrica.

A **tabela** a seguir apresenta os `métodos` mais importantes dos sets.



No fim do capítulo, a secção **Síntese** compara listas, tuplos, dicionários e conjuntos e ajuda-te a escolher a estrutura certa.


## Operações básicas

### `Criar um set`


```python
n_ips = {"192.168.0.1", "192.168.0.112", "192.168.0.233", "192.168.0.1"}
print(type(n_ips))
print(n_ips)
```

Também pode ser criado com a função `set()`, a partir de qualquer sequência (os duplicados são eliminados).

```python
nsets = set(range(10))
print(nsets)
print(set([1, 1, 2, 2, 3]))  # os duplicados desaparecem
print(type({}), type(set()))
```

E, a utilização do `for` permite **iterar sobre os elementos de um set**.

```python
for itens in nsets:
    print(itens, end = " ")
```

E mais uma vez a função `len()` é utilizada para determinar o `número de elementos` num set.

```python
print(len(nsets))
```

Os `operadores in e not in` também podem ser utilizados com `sets`.

```python
print(1 in nsets)
print(4 in nsets)
print(10 not in nsets)
print(3 not in nsets)
```

### `Método add()`

Permite adicionar um elemento ao `set`.

```python
nsets.add(11)
print(nsets)
```

### `Método discard()`

Permite `remover um` elemento do `set`. Se o elemento não existir, não acontece nada (ao contrário de `remove()`, que veremos mais abaixo e que dá erro).


```python
nsets.discard(0)
print(nsets)
```

### `Método clear()`

Permite `remover todos` os elementos do `set`, retornando um `set() vazio`.

```python
n_ips.clear()
print(n_ips)
```

O `set` também pode ser utilizado para `eliminar letras repetidas` de uma *string*.

```python
print(set("O rato roeu a roupa do Rei de Roma"))
```

### `Interseção`

Para a *interseção* utiliza-se o `operador &` ou o `método intersection()`.


```python
x = {1, 2, 3, 4 ,5, 7, 9}
y = {0, 2, 4, 6, 8, 10}

print(x & y)
print(x.intersection(y))
```

### `Diferença`

Para a *diferença* utiliza-se o `operador -` ou o `método difference()`.

```python
# Relativo a x
print(x - y)
print(x.difference(y))

# Relativo a y
print(y - x)
print(y.difference(x))
```

### `União`

Para *união* utiliza-se o `operador |` ou o `método union()`.


```python
print(x | y)
print(x.union(y))
z = x.union(y)
```

### `Diferença simétrica e subconjuntos`

- A **diferença simétrica** (`^` ou `symmetric_difference()`) devolve os elementos que estão num dos conjuntos, mas não nos dois.
- `issubset()` (ou `<=`) verifica se todos os elementos de um conjunto estão noutro; `issuperset()` (ou `>=`) faz o inverso.


```python
print(x ^ y)
print(x.symmetric_difference(y))

pares_pequenos = {2, 4}
print(pares_pequenos.issubset(y))   # True
print(pares_pequenos <= x)          # True
print(y.issuperset({0, 10}))        # True
```

### `Método remove()`

Este método `remove` o elemento especificado do set. Se o elemento não existir, dá um `KeyError` (por isso, `discard()` é mais seguro quando não temos a certeza).

```python
z.remove(10)
print(z)
```

### `Método pop()`

Este método remove e devolve um elemento **arbitrário** do set. Como os sets não têm ordem, não é possível escolher qual será removido.

```python
removido = z.pop()
print("Removido:", removido)
print(z)
```


## Síntese: que estrutura de dados usar?

Estudámos quatro estruturas de dados: listas (`list`), tuplos (`tuple`), dicionários (`dict`) e conjuntos (`set`). Em princípio, são em parte permutáveis.

Por exemplo, qualquer *lista* poderia ser simulada com um *dicionário*, usando o `índice` de cada item como `chave`. Contudo, algumas operações tornar-se-iam **muito complexas**: ao remover um item de uma lista, os índices dos itens seguintes são ajustados automaticamente; com um dicionário, teríamos de renumerar as chaves "à mão".

Usar listas em vez de dicionários (ou vice-versa) tem também *implicações no desempenho e na utilização de memória* de um programa. Uma discussão mais detalhada das `vantagens e desvantagens` de cada estrutura está *fora do âmbito desta formação*.

Outro exemplo: qualquer tuplo pode ser substituído por uma lista. Teríamos mais liberdade, mas perderíamos a garantia de *imutabilidade* que o tuplo oferece.

Abaixo, uma proposta de quando utilizar cada estrutura. É fruto da experiência pessoal em desenvolvimento e deve ser vista como uma recomendação, não como verdade absoluta: outros programadores podem ter opiniões diferentes.

| | Lista | Tuplo | Conjunto | Dicionário |
| --- | --- | --- | --- | --- |
| Sintaxe | `[1, 2]` | `(1, 2)` | `{1, 2}` | `{"a": 1}` |
| Ordenado | Sim | Sim | Não | Sim (ordem de inserção, Python 3.7+) |
| Mutável | Sim | Não | Sim | Sim |
| Duplicados | Sim | Sim | Não | Chaves únicas |
| Vazio | `[]` | `()` | `set()` | `{}` |
| Quando usar | coleção de itens do mesmo tipo, que muda | registo de dados que não muda | eliminar duplicados, testar pertença, operações de conjuntos | aceder a valores por nome (chave) |
