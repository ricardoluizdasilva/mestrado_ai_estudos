## Programação

A **`programação`** de computadores é o processo de escrever instruções para que um computador execute uma tarefa específica. É como dar ordens a um computador, mas numa linguagem que ele interprete e entenda.

- **Compilação e interpretação:** O código-fonte precisa ser traduzido para uma linguagem que o computador entenda. Isso pode ser feito por um compilador ou um interpretador.

## 5. Algoritmos

O primeiro passo para aprender programação não envolve necessariamente um computador, mas sim educar o nosso cérebro para entender a lógica por trás da resolução de problemas. Antes de transformar um algoritmo em código, é essencial compreender como estruturar as etapas necessárias para executar uma determinada tarefa.


### 5.1 O que é um algoritmo?

Um **`algoritmo`** é uma sequência finita de passos lógicos e organizados para resolver um problema. Como definido por Ziviani (2011), `é uma sequência finita de ações executáveis que visam obter uma solução para determinado tipo de problema`. De forma semelhante, Dasgupta, Papadimitriou e Vazirani destacam que `algoritmos são procedimentos precisos, não ambíguos, padronizados, eficientes e corretos`.





Embora frequentemente associado a programação, um algoritmo não precisa, necessariamente, ser implementado num computador. Ele pode ser seguido por um ser humano ou qualquer outro tipo de autómato. O mais importante é que ele descreva com clareza e precisão as etapas para alcançar um objetivo.


### 5.2 Características de um algoritmo

Um **`algoritmo`** deve possuir certas propriedades fundamentais para garantir sua eficiência:
- **Preciso (não ambíguo):** Cada instrução deve ter apenas uma interpretação possível.
- **Executável (eficaz):** Cada passo deve poder ser realizado na prática.
- **Determinístico:** Se as instruções forem seguidas corretamente, o resultado esperado será sempre o mesmo.
- **Finito:** O algoritmo deve terminar após um número limitado de passos.

Além disso, existem diversas maneiras de resolver um mesmo problema, ou seja, podem ser criados vários algoritmos diferentes para alcançar um mesmo resultado. A escolha da abordagem mais eficiente depende do contexto e da complexidade da tarefa a ser resolvida.

## 6. Entrada e Saída de dados (Input and Output)


### 6.1 Fundamentos de arquitetura básica de computadores




A arquitetura de um computador baseia-se num conjunto de componentes que interagem para processar e transmitir informações. A imagem apresentada ilustra o fluxo de dados dentro de um sistema computacional, destacando três elementos principais: Entrada, Processamento e Saída (Input, Processing, and Output).

**`Dispositivos de Entrada (Input)`**

Os dispositivos de entrada permitem que os utilizadores enviem dados para o computador, como por exemplo:
- **Teclado:** Utilizado para inserir textos e comandos.
- **Scanner:** Digitaliza documentos e imagens para o formato digital.
- **Microfone:** Capta áudio para gravação ou comunicação.
- **Câmara:** Captura imagens e vídeos.

Esses dispositivos enviam informações para a Unidade Central de Processamento (CPU), onde os dados serão processados.

**`CPU (Unidade Central de Processamento)`**
A `CPU` é o cérebro do computador. A sua função é processar dados e executar instruções fornecidas pelos dispositivos de entrada. Para isso, ela interage com diferentes tipos de memória:
- **RAM (Memória de Acesso Aleatório):** Armazena temporariamente os dados em uso, garantindo rapidez na execução de programas.
- **ROM (Memória Só de Leitura):** Contém informações essenciais para o funcionamento do sistema, como o `firmware`.
- **STORAGE (Armazenamento):** Dispositivos como discos rígidos e SSDs que mantêm os dados de forma permanente.

**`Dispositivos de saída (Output)`**
Os dispositivos de saída recebem as informações processadas e as apresentam ao utilizador. Alguns exemplos comuns são:
- **Ecrã:** Exibe gráficos e informações visuais.
- **Impressora:** Converte documentos digitais em papel.
- **Colunas:** Emitem sons e reproduzem áudios.

Esse fluxo contínuo de dados – da Entrada (`Input`) para o Processamento (`Processing`) e, por fim, para a Saída (`Output`) – é a base do funcionamento de qualquer sistema computacional.





Em **resumo**, a arquitetura de um computador baseia-se num conjunto de componentes que interagem para processar e transmitir informações. O funcionamento de um sistema computacional pode ser resumido em três elementos principais: **Input** (Entrada), **Processing** (Processamento), e **Output** (Saída). O **`Feedback`** representado na imagem refere-se ao processo pelo qual o sistema computacional avalia a saída gerada e a utiliza para ajustar, melhorar ou corrigir futuras execuções, podendo ser manual ou automática. Esse mecanismo garante que o sistema possa responder dinamicamente a diferentes condições e otimizar seu funcionamento. Esse processo pode envolver monitorização do desempenho, correção de erros, aprendizagem baseado em dados anteriores e aperfeiçoamento da execução de tarefas.





## 7. Operadores

Os **`operadores`** são símbolos que realizam operações sobre valores, permitindo a criação de expressões matemáticas e lógicas que produzem um resultado.

Por exemplo, na expressão `2 + 2`, o símbolo `+` é um **operador aritmético** que representa a soma, e o resultado dessa operação é `4`. Os operadores estabelecem relações entre valores para gerar um novo valor.




### 7.1 Tipos de operadores

Os **`operadores podem`** ser classificados em três categorias principais:

**Operadores aritméticos:** Realizam operações matemáticas básicas **(`+`, `-`, `*`, `/`, `%`, `**`)**.

| Operador  | Descrição       | Exemplo   |Resultado   |
| --------- | --------------- | --------- |----------- |
| +         | Adição          |  5 + 3    |  8         |
| -         | Subtração       | 10 - 4    |  6         |
| *         | Multiplicação   |  6 * 2    | 12         |  
| /         | Divisão         |  8 / 2    |  4.0       |
| %         | Módulo (resto)  | 10 % 3    |  1         |
| **        | Exponenciação   | 3 ** 2    |  9         | 



**Operadores relacionais:** Comparam valores e retornam um resultado lógico True ou False **(`==`, `!=`, `>`, `<`, `>=`, `<=`)**.

| Operador  | Descrição       | Exemplo   |Resultado   |
| --------- | --------------- | --------- |----------- |
| ==        | Igualdade       |  5 == 5   |  True      |
| !=        | Diferente       |  3 != 4   |  True      |
| >         | Maior que       |  10 > 5   |  True      |  
| <         | Menor que       |   4 < 7   |  True      |
| >=        | Maior ou igual  |  6 >= 6   |  True      |
| <=        | Menor ou igual  |  3 <= 2   |  False     | 




**Operadores lógicos:** Avaliam múltiplas condições para retornar um resultado lógico **(`and`, `or`, `not`)**.

| Operador  | Descrição       | Exemplo           |Resultado   |
| --------- | --------------- | ----------------- |----------- |
| not       | Negação         |  not True         |  False     |
| and       | Conjunção       |  True and False   |  False     |
| or        | Disjunção       |  True or False    |  True      |  



### 7.2 Ordem das operações aritméticas
Assim como na matemática, a ordem das operações segue uma hierarquia:
1. **Parênteses**: Expressões dentro de parênteses são resolvidas primeiro.
2. **Potenciação e Radiciação**: `**` (exponenciação) é avaliado antes das demais operações.
3. **Multiplicação, Divisão, Divisão inteira e Módulo** (`*`, `/`, `//`, `%`): têm a mesma precedência e são resolvidas da esquerda para a direita.
4. **Adição e Subtração**: Resolvidas por último, também da esquerda para a direita.

### 7.3 Ordem das operações lógicas
A ordem de execução das operações lógicas, também conhecida como precedência de operadores lógicos, define a sequência em que as operações são realizadas numa expressão lógica. Essa ordem é crucial para garantir que o resultado da expressão seja correto e previsível.

A hierarquia das operações lógicas segue a seguinte ordem:
1. **`not` (Negação)**: Primeiro, inverte o valor lógico.
2. **`and` (Conjunção)**: Depois, avalia as expressões que precisam ser verdadeiras.
3. **`or` (Disjunção)**: Por último, avalia as expressões que podem ser verdadeiras ou falsas.


#### 7.3.1 Operador lógico `AND`

O Operador **`AND`** resulta num valor `VERDADEIRO se os dois valores de entrada da operação forem VERDADEIROS`, caso contrário o resultado é `FALSO`.

|    VALOR 1     |    VALOR 2    | Operação AND |
|----------------|---------------|--------------|
|   **TRUE**     |   **TRUE**    |   **TRUE**   |
| TRUE           | FALSE         |  FALSE       |
| FALSE          | TRUE          |  FALSE       |
| FALSE          | FALSE         |  FALSE       |


#### 7.3.2 Operador lógico `OR`

O Operador **`OR`** resulta num valor `VERDADEIRO se ao menos UM dos dois valores de entrada da operação for VERDADEIRO`, caso contrário o resultado é `FALSO`.

|    VALOR 1     |    VALOR 2    | Operação OR  |
|----------------|---------------|--------------|
|   **TRUE**     |   **TRUE**    |   **TRUE**   |
|   **TRUE**     | FALSE         |   **TRUE**   |
| FALSE          |   **TRUE**    |   **TRUE**   |
| FALSE          | FALSE         |  FALSE       |


#### 7.3.3 Operador lógico `NOT`

O Operador `NOT` é o único operador que recebe como `entrada apenas um valor`, e sua função é simplesmente `inverter os valores`.
Ou seja, `se o valor de entrada for VERDADEIRO, o resultado será FALSO` e se o `valor de entrada for FALSO, o resultado será VERDADEIRO`.

| VALOR ENTRADA 1 |  Operação NOT  |
|-----------------|----------------|
|      TRUE       |   **FALSE**    |
|      FALSE      |   **TRUE**     |


#### 7.3.4 Operador lógico `NAND`

O Operador `NAND` é o contrário do operador `AND`, ou seja, resulta em `VERDADEIRO`,` se ao menos um dos dois valores for FALSO`. Ou seja, é o operador `AND seguido do operador NOT`.

|    VALOR 1     |    VALOR 2    | Operação NAND  |
|----------------|---------------|----------------|
| TRUE           | TRUE          | FALSE          |
|   **TRUE**     |   **FALSE**   |   **TRUE**     |
|   **FALSE**    |   **TRUE**    |   **TRUE**     |
|   **FALSE**    |   **FALSE**   |   **TRUE**     |


#### 7.3.5 Operador lógico `NOR`

O Operador `NOR` é o `contrário do operador OR`, ou seja, `resulta em VERDADEIRO, se os dois valores forem FALSO`. Ou seja, é o operador `OR seguido do operador NOT`.

|    VALOR 1     |    VALOR 2    | Operação NOR   |
|----------------|---------------|----------------|
| TRUE           | TRUE          | FALSE          |
| TRUE           | FALSE         | FALSE          |
| FALSE          | TRUE          | FALSE          |
|   **FALSE**    |   **FALSE**   |   **TRUE**     |


#### 7.3.6 Operador lógico `XOR`

O Operador `XOR` (OU Exclusivo) é uma `variação` interessante do operador `OR`, ele `resulta em VERDADEIRO se apenas um dos valores de entrada for VERDADEIRO`, ou seja, `apenas se os valores de entrada forem DIFERENTES`.

|    VALOR 1     |    VALOR 2    | Operação XOR   |
|----------------|---------------|----------------|
| TRUE           | TRUE          | FALSE          |
|    **TRUE**    |   **FALSE**   |   **TRUE**     |
|    **FALSE**   |   **TRUE**    |   **TRUE**     |
| FALSE          | FALSE         | FALSE          |



### 7.4 Curiosidades (operadores lógicos) 

Algumas linguagens oferecem operadores lógicos para o nível de bit (também chamado de operadores bitwise). Ou seja, podemos fazer operações lógicas com os bits de dois números.

Cada linguagem de programação possui sua própria forma de representar os operadores lógicos, mas a simbologia mais comum inclui `and`, `or` e `not`.

Além desses operadores tradicionais, algumas linguagens oferecem operadores **bitwise**, que realizam operações diretamente nos bits dos números fornecidos.

Em Python (tal como em Java ou C), estes operadores são, por exemplo:
- **`&`** (AND bitwise)
- **`|`** (OR bitwise)

Essas operações lógicas são realizadas com os bits dos números de entrada.

**Assim:**

Convertemos o número 10 e o número 7 para binário:
- 10 = 1010 em binário
- 7 = 0111 em binário

**Depois** realizamos as operações lógicas com cada bit dos dois números.
Da direita para a esquerda aplicamos as operações lógicas para cada bit.
- 0 E 1 = 0
- 1 E 1 = 1
- 0 E 1 = 0
- 1 E 0 = 0

**Logo,**
 
- 10 & 7 = 0010 = 2 em números decimais.
- O mesmo para o operador OU:
- 0 OU 1 = 1
- 1 OU 1 = 1
- 0 OU 1 = 1
- 1 OU 0 = 1
 
**Logo,**
- 10 | 7 = 1111 = 15

#### Exemplo com os números 10 e 7 em Python

```python
x = 10  # 1010 em binário
y = 7   # 0111 em binário

print(x & y)  # AND bitwise
print(x | y)  # OR bitwise
```

Os operadores bitwise são úteis em programação de sistemas embebidos, criptografia e otimização de código de baixo nível.

**`Nota importante:`** Os operadores serão abordados com mais detalhe no capítulo 8 de introdução ao Python.


## 8. Variáveis e Tipos de Dados

### 8.1 Variáveis

Programas de computador utilizam os recursos de hardware mais básicos para executar `algoritmos`.

Enquanto o processador executa os cálculos, a `memória` é responsável por armazenar dados e servi-los ao processador. 

O recurso utilizado nos programas para escrever e ler dados da memória do computador é conhecido como **`variável`**, que é simplesmente um espaço na memória o qual reservamos e atribuímos um nome.

Por exemplo, podemos criar uma **`variável`** chamada `idade` para armazenar a idade de uma pessoa.

Imagina uma **`variável`** como uma gaveta com uma etiqueta num armário.



A este espaço alocado na `memória` chamamos de **`variável`**, porque o valor armazenado neste espaço de memória pode ser alterado ao longo do tempo, ou seja, o valor ali alocado é “variável” ao longo do tempo.


Diferente das **`constantes`**, que é um espaço reservado na memória para armazenar um valor que não muda com o tempo. Neste sentido, uma `constante` é um espaço de memória que armazena um `valor fixo`, que não pode ser alterado durante a execução do programa. O `valor de uma constante` é definido no momento da sua declaração e permanece o mesmo durante todo o programa.

As **`constantes`** são utilizadas para armazenar valores que não devem ser modificados, como valores matemáticos (pi), configurações de programas ou dados que precisam permanecer inalterados.

Por exemplo, o valor `PI (3.14159265359…)` que nunca vai mudar, ou o `número de dias na semana (7)`, ou ainda a `velocidade da luz no vácuo (299 792 458 m/s)`.




### 8.2 Tipos de Dados

Em programação, os dados são classificados em duas categorias principais: `tipos primitivos e tipos personalizados`.




**Tipos Primitivos**
- **Inteiro (int):** Representa números inteiros, positivos ou negativos, sem casas decimais. **`Exemplo:`** idade (25), número de alunos (30).
- **Real (float/double):** Representa números reais, com casas decimais. **`Exemplo:`** peso (65.5 kg), altura (1.75 m).
- **Lógico (boolean):** Representa valores lógicos, verdadeiro (true) ou falso (false). Usado para condições e decisões. **`Exemplo:`** lâmpada acesa (true), utilizador autenticado (false).
- **Texto (string):** Representa sequências de caracteres. **`Exemplo:`** nome ("João"), endereço ("Rua A, 123").


**Tipos Personalizados**
- São criados pelo programador para representar entidades complexas. Uma classe é um exemplo, agrupando dados e comportamentos relacionados.
- Exemplo: a **`classe "Carro`** pode ter atributos como `marca`, `modelo`, `cor` e `métodos` como `ligar()`, `acelerar()`.
- Diferente dos tipos primitivos, que são nativos da linguagem, os tipos personalizados permitem modelar o mundo real dentro do código.

O conceito de estruturas de dados e classes é bem mais complexo que isto, por exemplo classes têm operações além de atributos, mas aqui é importante apenas frisar que também são tipos de dados. Diferente dos tipos de dados primitivos que já são implementados internamente pelas linguagens de programação, esses tipos de dados são criados pelo programador.



## 9. Estruturas de controlo

Em programação, as estruturas de controlo são ferramentas fundamentais que determinam a ordem em que as instruções são executadas num programa. Elas permitem que o programa tome decisões, repita ações e organize o fluxo de execução de forma lógica e eficiente.

Os principais tipos de estrutura de controlo são:
- **Estruturas de decisão (ou condicionais):** Permitem que o programa escolha entre diferentes caminhos de execução com base em condições lógicas.
- **Estruturas de repetição (ou loops):** Permitem que o programa execute um bloco de código repetidamente até que uma condição seja satisfeita.
- **Estruturas de desvio:** Permitem que o fluxo de execução do programa seja desviado para outro ponto do código.


### 9.1 Estruturas de decisão

Em programação, as **`estruturas de decisão`** são ferramentas essenciais que permitem que um programa tome diferentes caminhos de execução com base em condições específicas. Em outras palavras, elas permitem que o **`programa decida qual bloco de código executar`**, dependendo se uma **`determinada condição é satisfeita`**.

**Conceitos fundamentais**

- Condições: As estruturas de decisão dependem de condições, que são expressões que resultam num valor booleano: verdadeiro (true) ou falso (false).
Essas condições geralmente envolvem comparações (por exemplo, se um número é maior que outro) ou avaliações lógicas (por exemplo, se duas condições são verdadeiras).

- Fluxo de controlo: As estruturas de decisão controlam o fluxo de execução do programa, determinando quais instruções serão executadas e quais serão ignoradas.


**Tipos comuns de estruturas de decisão:**

- **if (se):** Executa um bloco de código somente se a condição especificada for verdadeira.
**`Exemplo: Se a idade for maior ou igual a 18, exibir "Maior de idade".`**
  
- **if-else (se-senão)**: Executa um bloco de código se a condição for verdadeira e outro bloco se a condição for falsa.
**`Exemplo: Se a idade for maior ou igual a 18, exibir 'Maior de idade'; senão, exibir "Menor de idade".`**

- **switch (escolha):** Permite escolher entre múltiplos blocos de código com base no valor de uma variável ou expressão. É útil quando há muitas opções a serem consideradas. O switch funciona da mesma forma que o switch das redes de computadores, só que em vez de enviar um pacote de dados para uma determinada porta, vamos enviar o fluxo do algoritmo para um determinado ponto do código.
 Em Python, o equivalente é a instrução `match/case` (Python 3.10 ou superior), que veremos no capítulo 10.

**Importância das estruturas de decisão:**
- São cruciais para criar programas flexíveis e adaptáveis.
- Permitem que os programas respondam a diferentes situações e entradas de dados.
- São usadas numa ampla variedade de aplicações, desde simples verificações de entrada de dados até complexos sistemas de controlo.

Em resumo, as estruturas de decisão são a base da lógica condicional em programação, permitindo que os programas tomem decisões e executem ações apropriadas com base em diferentes cenários.

`Abaixo` um exemplo de código Python para o `if-else`.


```python
idade = 20  # experimenta alterar este valor

if idade >= 18:
    print("Maior de idade")
else:
    print("Menor de idade")
```

### 9.2 Estrutura de repetição

As **estruturas de repetição**, também conhecidas como `ciclo ou laços (loops)`, são ferramentas fundamentais na programação que permitem `executar um bloco de código repetidamente até que uma determinada condição seja satisfeita`. Elas são essenciais para automatizar tarefas repetitivas e processar grandes volumes de dados.

**Tipos de estruturas de repetição**

Existem três tipos principais de estruturas de repetição:

- **for:** Utilizado quando se conhece o número exato de repetições que serão necessárias.
Em linguagens como C ou Java, consiste em três partes: inicialização, condição e incremento/decremento. Em Python, o `for` percorre diretamente os elementos de uma sequência (por exemplo, `for i in range(1, 11)`).
**`Exemplo: imprimir os números de 1 a 10.`**
  
- **while:** Utilizado quando o número de repetições é desconhecido e depende de uma condição.
O bloco de código é executado enquanto a condição for verdadeira. É importante garantir que a condição eventualmente se torne falsa para evitar loops infinitos.
**`Exemplo: ler dados de um ficheiro até encontrar o final.`**

- **Loop "do-while":** Semelhante ao loop `while`, mas garante que o bloco de código seja executado pelo menos uma vez.
A condição é verificada após a execução do bloco de código. Python não tem `do-while`; simula-se com `while True:` e `break`.

**Importância das estruturas de repetição:**
- **Automatização de tarefas:** Permitem executar tarefas repetitivas de forma eficiente, poupando tempo e esforço.
- **Processamento de dados:** Facilitam o processamento de grandes volumes de dados, como listas e arrays.
- **Controlo de fluxo:** Permitem controlar o fluxo de execução do programa, repetindo ações até que uma condição seja satisfeita.

Em resumo, as estruturas de repetição são ferramentas poderosas que permitem criar programas mais eficientes, flexíveis e capazes de lidar com tarefas complexas.


### 9.3 Estrutura de desvio

Em programação, as estruturas de desvio são mecanismos que alteram o fluxo normal de execução de um programa, permitindo que o código salte para diferentes partes com base em determinadas condições. São cruciais para criar programas flexíveis e adaptáveis.

**Tipos principais de estruturas de desvio**

- **break:** Utilizado para interromper a execução de um `loop (como for ou while)` ou de uma `estrutura switch`. Quando `break` é encontrado, o programa `sai imediatamente do loop ou do switch`, continuando a execução na próxima instrução após a estrutura interrompida.

- **continue:** Utilizado `dentro de loops para saltar a iteração atual e ir para a próxima`. Quando `continue` é encontrado, o código restante dentro da iteração atual é ignorado, e o `loop prossegue` para a próxima iteração.

- **return:** Utilizado em funções para encerrar a execução da função e retornar um valor `(se a função for projetada para retornar algo)`. Quando `return` é encontrado, o `programa volta para o ponto onde a função foi chamada`.

- **goto:** Menos comum, específico de determinadas linguagens de programação (não existe em Python) e geralmente desaconselhado. `Permite saltar para um ponto específico do código marcado por um rótulo`. O uso excessivo de `goto pode tornar o código difícil` de entender e manter, levando a um `código espaguete`.

**Importância das estruturas de desvio:**
- **Controlo de fluxo preciso:** Permitem que os programadores controlem o fluxo de execução de um programa de forma precisa.
- **Tratamento de erros:** Permitem lidar com situações excecionais ou erros de forma eficiente.
- **Otimização de código:** Podem ser usadas para otimizar o desempenho do programa, evitando a execução de código desnecessário.

Em resumo, as estruturas de desvio são ferramentas poderosas que permitem criar programas mais complexos e eficientes.


## 10. Vetores e Matrizes

Em programação, **vetores e matrizes** são `estruturas de dados` usadas para armazenar múltiplos valores organizados de forma eficiente. Eles são amplamente utilizados em computação para representar listas de valores, conjuntos de dados e até mesmo estruturas mais complexas, como imagens ou grafos.

**Vetores:** Também conhecido como **array unidimensional**, é uma sequência ordenada de elementos do mesmo tipo.
Cada elemento do `vetor` é identificado por um `índice`, que geralmente começa em 0.
É como uma `lista linear`, onde cada posição `armazena um valor`.

**`Exemplo:`**
Um `vetor para armazenar as idades` de um grupo de pessoas: [25, 30, 22, 28].
Neste exemplo, `25` seria o `índice 0`, `30` o `índice 1` e `assim por diante`.

**Matrizes:** Ou **array multidimensional**, é uma coleção de `elementos organizados em linhas e colunas`. Cada elemento é identificado por `dois índices`: um para a `linha` e outro para a `coluna`. É como uma tabela, onde cada célula armazena um valor.

**`Exemplo:`**
Uma `matriz` para representar as `notas de alunos` em diferentes disciplinas:

`[[8.5, 9.0, 7.5],
 [6.0, 7.0, 8.0],
 [9.5, 8.5, 9.0]]`
 
Neste exemplo, a `primeira linha` representa as `notas do primeiro aluno`, a `segunda linha` as `notas do segundo aluno`, e assim sucessivamente.


Os **`vetores`** e **`matrizes`** são importantes porque:
- **Organização de dados:** Vetores e matrizes permitem organizar grandes quantidades de dados de forma estruturada.
- **Acesso eficiente:** Permitem aceder a elementos individuais de forma rápida e direta, usando os seus índices.
- **Manipulação de dados:** Facilitam a realização de operações em conjuntos de dados, como busca, ordenação e cálculos.
- **Aplicações:** São amplamente utilizados em diversas áreas, como processamento de imagens, jogos, simulações e análise de dados.

Em resumo:
- Vetores são como listas lineares, enquanto matrizes são como tabelas.
- Ambos permitem armazenar coleções de elementos do mesmo tipo.
- São ferramentas essenciais para organizar e manipular dados em programação.



## 11. Funções e Procedimentos

Em programação, **funções e procedimentos** são `blocos de código reutilizáveis` que realizam `tarefas específicas`. **`As funções e os procedimentos`** ajudam a `organizar o código`, torná-lo mais legível e `evitar a repetição de código`.

- **Funções:** Uma função é um bloco de código que executa uma tarefa específica e pode retornar um valor. São projetadas para realizar cálculos, manipulações de dados ou outras operações que produzem um resultado.

`As suas principais características são:`
- Recebem parâmetros (entradas) opcionais.
- Executam um conjunto de instruções.
- Retornam um valor (saída) usando a palavra-chave "return".

**`Exemplo:`**
Uma função que calcula a soma de dois números e retorna o resultado.


- **Procedimentos:** Ou sub-rotina, é um bloco de código que executa uma tarefa específica, mas não retorna um valor. Os procedimentos são usados para realizar ações, como exibir mensagens, modificar variáveis ou executar outras operações que não exigem um resultado direto.

`As suas principais características são:`
- Recebem parâmetros (entradas) opcionais.
- Executam um conjunto de instruções.
- Não retornam um valor diretamente.

**`Exemplo:`**
Um procedimento que exibe uma mensagem de boas-vindas no ecrã.

As principais diferenças entre `funções` e `procedimentos` são:
- **Retorno de valor:** As `funções` retornam um valor. Os `procedimentos` não retornam um valor diretamente.
- **Utilização:** As `funções` são usadas para cálculos e operações que produzem um resultado. Os `procedimentos` são usados para ações e tarefas que não exigem um resultado direto.

**`Por que são relevantes?`**
- **Reutilização de código:** Funções e procedimentos permitem reutilizar blocos de código, evitando a repetição de instruções.
- **Organização do código:** Ajudam a dividir o código em módulos menores e mais fáceis de entender.
- **Legibilidade do código:** Tornam o código mais legível e fácil de manter.
- **Modularidade:** Facilitam a criação de programas complexos, dividindo-os em partes menores e independentes.

Em muitas linguagens de programação modernas, a distinção entre funções e procedimentos é ténue. Em algumas linguagens como C e Java, os procedimentos são funções que não retornam nada (void).
