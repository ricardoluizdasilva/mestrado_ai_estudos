# Capítulo 13 – Tuplos

Um tuplo (`tuple`) é uma estrutura de dados que agrupa vários elementos, tal como a lista. A diferença é que os tuplos são `imutáveis`: depois de criados, não podem ser modificados.

Um tuplo escreve-se entre parênteses `()`, com os elementos separados por vírgulas `,`, e pode ser atribuído a uma variável. Um tuplo pode conter elementos de qualquer tipo, incluindo outros tuplos ou listas.

Abaixo, alguns **exemplos** de `tuplos` (os dados são fictícios):

```python
endereco = (1000, "Lisboa", "Rua do Exemplo 123, 1.º andar", "+351 210 000 000")
print(endereco)
aluno = ("Pedro", "Silva", 123456, "Consultor", "pedro.silva@exemplo.pt", endereco)
print(aluno)
tup1 = (12312, "absbsb", [1, 2, 3, 4], ("a", "b", "c"), "fim")
print(tup1)
linux = ("red hat", " suse", "kali", "ubuntu", "fedora")
print(linux)
print(type(endereco), type(aluno), type(tup1), type(linux))
```

Um tuplo não pode ser modificado.

A grande diferença entre listas e tuplos é esta: as listas podem ser modificadas e os tuplos não. Os `tuplos são imutáveis`, ou seja, depois de criado o tuplo:
- não se pode acrescentar um elemento;
- não se pode apagar um elemento;
- não se pode alterar o valor de um elemento.

Isto não impede que a variável passe a referir-se a outro tuplo (segunda célula abaixo), mas não existem métodos como `append()` ou `remove()`.


```python
# ⚠️ Esta célula gera um erro de propósito (TypeError). Lê a mensagem e continua.
tup = (1, 2, "a")
print(tup[2])
tup[2] = "b"
```

```python
tup = (1, 2, 3)
print(tup)
tup = ("a", "b", "c", "d")
print(tup)
```

Os elementos de um tuplo podem ser acedidos como numa lista, conforme o **exemplo** abaixo:

```python
linux = ("red hat", " suse", "kali", "ubuntu", "fedora")
print(linux[0:2])
print(linux[-1])
print(linux[:-2])
print(linux[:])
```

Como já foi mencionado, os tuplos são semelhantes às listas. São muito utilizados para representar registos de dados.

Uma regra prática para escolher entre listas e tuplos:
- as listas são usadas para coleções de elementos do mesmo tipo, que podem crescer ou mudar (por exemplo, uma lista de nomes);
- os tuplos são usados para agrupar dados diferentes que descrevem uma mesma coisa e que não mudam (um registo).

Considera os seguintes dados como exemplos de tuplos:

- `endereco = (codigo_postal, cidade, rua, telefone)`
- `posicao = (coordenada_x, coordenada_y)`
- `data = (ano, mes, dia)`

> **Atenção:** um tuplo com um único elemento precisa de uma vírgula: `(5,)`. Sem a vírgula, `(5)` é apenas o número 5 entre parênteses.


```python
a = (5)
b = (5,)
print(type(a), type(b))
```

## Desempacotamento (unpacking)

Os elementos de um tuplo podem ser atribuídos a várias variáveis de uma só vez. É uma das utilizações mais comuns dos tuplos (já a usámos na atribuição múltipla `a, b = 10, 5`, no capítulo 4) e permite, por exemplo, trocar o valor de duas variáveis sem variável auxiliar.


```python
posicao = (3, 7)
x, y = posicao          # desempacotamento
print("x =", x, "e y =", y)

a, b = 1, 2
a, b = b, a             # troca de valores
print(a, b)
```

## Operações básicas

### `Criar um tuplo com input()`
Podemos utilizar a função `input()` para obter valores e construir um tuplo, conforme o exemplo da célula abaixo.


```python
nome = input("Introduza o seu nome: ")
fone = input("Introduza o seu telefone: ")
idade = int(input("Introduza a sua idade: "))

aluno = (nome, fone, idade)

print(aluno)
```

### `Aceder a um elemento do tuplo`

Tal como numa `list`, os elementos de um `tuple` têm um índice, indicado entre `[]` e que começa em 0. Os índices negativos e o `slicing` funcionam exatamente como nas listas.


```python
endereco = (1000, "Lisboa", "Rua do Exemplo 123, 1.º andar", "+351 210 000 000")
aluno = ("Pedro", "Silva", 123456, "Consultor", "pedro.silva@exemplo.pt", endereco)

print(endereco[0])
print(aluno[1])
print(aluno[5])
print(aluno[5][2])
print(aluno[-1])
print(endereco[-3])
print(endereco[2])
```

A utilização do `for` permite **percorrer os elementos de um tuplo**.


```python
for e in aluno:
    print(e)
```

### `Verificar valores de um tuplo`

O operador `in` permite verificar se um determinado item **está presente** num tuplo; o operador `not in` faz o inverso.


```python
print("Pedro" in aluno)
print("José" in aluno)
print("Porto" in endereco)
print("Beja" not in endereco)
```

### `Verificar o tamanho de um tuplo`

Para saber o **tamanho** de um tuplo, ou seja, o **número de elementos**, utilizamos a já conhecida `função len()`.

```python
print(len(aluno))
print(len(endereco))
```

### `Apagar um tuplo`

Sabemos que um **tuplo é imutável** e que não é possível remover itens. Contudo, a variável pode ser apagada com a `palavra-chave del`. Depois disso, a variável deixa de existir, pelo que a segunda linha da célula abaixo dá um erro.


```python
# ⚠️ Esta célula gera um erro de propósito (NameError). Lê a mensagem e continua.
del aluno
print(type(aluno))
```

### `Método count()`

Este método devolve o `número de vezes` que um determinado valor ocorre num tuplo.

No **exemplo**, o tuplo é criado a partir de um `range()` com a função `tuple()`.

```python
numeros = tuple(range(1, 40, 3))
print(numeros)
print(numeros.count(7))
```

### `Método index()`

O método `index()` procura um determinado elemento e **devolve o seu índice** (a primeira ocorrência).


```python
print(numeros.index(19))
```

### `Conversão`

Já vimos funções de conversão, como `int()`, que converte uma string num inteiro.

Também é possível converter uma lista num tuplo, com `tuple()`, e vice-versa, com `list()`.

**Exemplos:**

```python
l = [1, 2, "a", 2.3]
t = tuple(l)
print(t)
l = list(t)
print(l)
```

### `Função dir()`

A função `dir()` permite ver todos os `atributos e métodos disponíveis` num objeto, por exemplo numa lista e num tuplo. Repara como o tuplo tem muito menos métodos (só `count` e `index`), por ser imutável.

```python
numeros_primos = [2, 3, 5, 7, 11, 13, 17, 19]
quadrados_perfeitos = tuple([x ** 2 for x in numeros_primos])
print("Métodos Lista", dir(numeros_primos))
print("Métodos Tuple", dir(quadrados_perfeitos))
```
