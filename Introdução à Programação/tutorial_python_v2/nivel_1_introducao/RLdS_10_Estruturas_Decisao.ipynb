# Capítulo 10 – Estruturas de decisão

Até agora, vimos instruções de código a serem executadas uma a uma até que o programa termine. Contudo, é frequente que partes do programa só sejam executadas de acordo com uma determinada condição.

Imagina a seguinte situação: `Amanhã, iremos à praia se não chover. Caso contrário, iremos ao centro comercial fazer compras e almoçar. À noite iremos jantar com amigos`.

A descrição do texto acima não é totalmente precisa, tal como as tarefas que serão executadas. Esta é uma descrição típica em linguagem natural. É frequente haver necessidade de informação adicional e mais contextualizada, a qual poderá ser obtida através de uma conversa, por **exemplo**. Uma conversão para pseudocódigo ficaria assim:


```text
Se amanhã chover:
    - Ir ao centro comercial
    - Fazer compras
    - Almoçar
Caso contrário:
    - Ir à praia

À noite, jantar com amigos.
```

A **tomada de decisão** é necessária quando se pretende executar parte de um código apenas se uma `determinada condição for satisfeita`. Este é um conceito muito importante, pois representa a capacidade lógica e analítica na execução de instruções.





É importante relembrar o conceito dos operadores relacionais que são utilizados para comparar valores (de qualquer tipo). O resultado de uma expressão relacional será um resultado `booleano` (`True` ou `False`).

| Símbolo   | Descrição        | Exemplo                |Significado                       |
| --------- | ---------------- | ---------------------- |--------------------------------- |
| >         | Maior que        | X > Y                  |`X` é maior que `Y` ?             |
| >=        | Maior ou igual   | X >= Y                 |`X` é maior que ou igual a `Y` ?  |
| <         | Menor            | X < Y                  |`X` é menor que `Y` ?             |  
| <=        | Menor ou igual   | X <= Y                 |`X` é menor que ou igual a `Y` ?  |
| ==        | Igual            | X == Y                 |`X` é igual a `Y` ?               |
| !=        | Diferente        | X != Y                 |`X` é diferente de `Y` ?          | 

É importante relembrar que o sinal `=` é utilizado para atribuir um valor a uma variável, enquanto o sinal `==` é usado para fazer comparações de igualdade, que retornará `True` se os valores forem iguais ou `False` se forem diferentes.

Essas `condições` podem ser utilizadas de várias maneiras, e são frequentemente utilizadas em instruções `condicionais (if)` e `loops`.


## Condição básica

Em Python as instruções `if`, `elif` e `else` são usadas para auxiliar na programação da tomada de decisões. Abaixo a representação da sintaxe:

```Python
if <condição_1>:
   instruções_1
   ...
elif <condição_2>:
   instruções_2
   ...
else:
   instruções_3
   ...
```

A indentação e os espaços têm significado em Python! Por isso, não se pode indentar de forma arbitrária dentro de um programa.

Neste exemplo, o programa executa a primeira linha `if <condição_1>:`. Se a expressão for avaliada como `True`, as instruções dentro do `bloco if` *`(instruções_1)`* serão `executadas`.

Caso a expressão da primeira linha retorne `False`, o programa irá passar para a linha `elif <condição_2>`. Se for `True`, as instruções do bloco `elif` *`(instruções_2)`* serão executadas. Se for falso, as instruções do bloco `else` *`(instruções_3)`* serão executadas.

É importante relembrar que a **`indentação`** define os limites dos blocos de código. As instruções que estão juntas são chamadas de blocos de código. Ao contrário de outras linguagens, que usam chavetas `{}` ou palavras-chave para marcar o início e o fim de um bloco, em Python não se usam marcadores especiais. Para tal, a indentação deve ser utilizada.

Vejamos alguns **exemplos práticos** com `if-elif-else`.

````python
# Exemplo prático de if-elif-else
# Entrada de dados
x = int(input("Introduza o 1º número: "))
y = int(input("Introduza o 2º número: "))

# Testa as diferentes condições
if x > y:
    print("O 1º número é maior do que o 2º")
elif x == y:
    print("Ambos os números são iguais")
else:
    print("O 2.º número é maior do que o 1º")

````

No `exemplo acima` a comparação é feita com números inteiros. O exemplo pode ser estendido a outros tipos de dados.

Como por **exemplo**, comparação entre `strings`.

````python
# Exemplo prático de if-elif-else
# String
nome = "Darth Vader"
if nome == "Darth Vader":
    print("I am your father")
else:
    print("I am not your father!")
````

Assim como em lógica, em Python (e noutras linguagens de programação) a condição pode ser feita pela negação, ou seja, com o `not`. Neste caso, há que ter em conta alguns aspetos importantes (recorda o capítulo 8):
- uma `string vazia` (`""`) é avaliada como `False`; qualquer outra string é considerada `True`;
- o número `0` é avaliado como `False`; qualquer outro número é considerado `True`;
- o mesmo acontece com listas, tuplos e dicionários vazios e com `None`, que são avaliados como `False`.

**Exemplos:**


````python
# Como a string está vazia (avaliada como False), "not senha" é True
senha = ""
if not senha:
    print("Senha inexistente!")
else:
    print("A senha existe!")

# Como 0 é avaliado como False, "not num" é True
num = 0
if not num:
    print("Número inexistente!")
else:
    print("O número existe!")
````

As **`premissas`** acima podem ser validadas através das seguintes instruções em Python:


````python
print(bool(""))
print(bool("A-2msg-493xxx"))
print(bool(0))
print(bool(5))
print(bool(-8))
print(bool(3.38))
print(bool(5+6j))
````

## Múltiplas condições

O que acontece quando há mais de duas alternativas? Por exemplo, se a variável nome contém os valores *João* ou *Maria* ou *António*?

Uma condição só pode ter dois resultados, `True` ou `False`. Contudo, é possível verificar várias condições uma após a outra.

```python
if condição_1:
   instruções_1
   ...
elif condição_2:
   instruções_2
   ...
elif condição_3:
   instruções_3
   ...
elif condição_4:
   instruções_4
   ...
else:
   instruções_5
   ...
```

Neste caso, as condições são verificadas em ordem, uma após a outra. Se a primeira condição for `True`, então o respetivo bloco de código será executado. Caso contrário, passará para a próxima condição, que é avaliada e, se for verdadeira, o respetivo bloco é executado. Se nenhuma das condições for satisfeita, é executado o bloco do `else`, ou seja, o bloco do `else` só é executado se nenhuma das condições anteriores for `True`. Assim que um bloco é executado, nenhuma outra condição é verificada.

Nas múltiplas condições a ordem das condições é crucial na lógica do programa.

**Exemplo:**


````python
# Reflete: o que acontece se a temperatura for 35? A mensagem "Está muito calor" alguma vez aparece?
temperatura = int(input("Qual é a temperatura atual (em Celsius)? "))

if temperatura > 20:
    print("Está calor")
elif temperatura > 30:
    print("Está muito calor")
else:
    print("Está frio")
````

Com 35 °C aparece "Está calor": a **primeira** condição verdadeira ganha e as restantes já não são testadas, por isso "Está muito calor" nunca é mostrado. Para corrigir, a condição mais restritiva (`> 30`) tem de vir primeiro:


````python
temperatura = 35

if temperatura > 30:
    print("Está muito calor")
elif temperatura > 20:
    print("Está calor")
else:
    print("Está frio")
````

## Nesting

São muitas as ocasiões em que há necessidade de se verificar várias condições, uma após a outra dentro do bloco de código dessa condição. A isto chamamos de `condições nested` (`nested conditions` ou `nesting`). Ou seja, um `if` pode conter um ou mais `if` dentro do mesmo **bloco de código**. 


````python
# Analisa o código
temperatura = int(input("Qual é a atual temperatura (em Celsius): "))
chuva = True
vento = False

if temperatura > 20:
    print("Está calor")
    if chuva:
        print("Está calor e chuvoso. Viva a primavera!")
        if vento:
            print("Está calor, chove e está ventoso!")
        else:
            print("Está calor, chove e não há vento ou brisa!")
    else:
        print("Está calor e seco! Nada agradável!")
        if vento:
            print("Está calor, seco e ventoso. Bom tempo para velejar!")
        else:
            print("Está calor, seco e não há vento ou brisa. Vamos à praia!")
else:
    print("Está frio. A chuva e o vento pouco importam. Vou ver um filme!")
````

## Expressão condicional
O Python suporta ainda outra forma de decisão, a expressão condicional (*conditional expression*, também conhecida como operador ternário). A sintaxe é:

```python
<expr1> if <condição> else <expr2>
```

A diferença para a instrução `if` normal é que a expressão condicional não controla diretamente o fluxo do programa: funciona como um operador que **produz um valor**. A `<condição>` é avaliada primeiro. Se for `True`, o resultado é `<expr1>`; se for `False`, o resultado é `<expr2>`.

Repara que a ordem de leitura não é muito óbvia: a expressão do meio é a primeira a ser avaliada e, com base no resultado, é escolhida uma das outras duas.

Mais um **exemplo** de uma `expressão condicional`.


````python
idade = 17
print("Menor de idade bebe leite!" if idade < 18 else "Maior de idade pode beber cerveja!")
````

## Escolha múltipla com `match/case`

Desde o Python 3.10 existe a instrução `match/case`, equivalente ao `switch` de outras linguagens referido na introdução à lógica de programação. Compara um valor com vários padrões e executa o bloco do primeiro que coincidir. O padrão `_` funciona como "caso contrário".


````python
dia = 6

match dia:
    case 1 | 2 | 3 | 4 | 5:
        print("Dia útil")
    case 6 | 7:
        print("Fim de semana")
    case _:
        print("Dia inválido")
````
