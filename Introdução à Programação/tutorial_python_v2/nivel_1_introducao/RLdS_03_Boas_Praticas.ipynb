## Sintaxe
O Python foi criado para ser uma linguagem simples e legível. Por isso, muitos iniciantes e especialistas das mais diversas áreas adotaram esta linguagem de programação, pela sua facilidade de uso e pela simplicidade e clareza da sua sintaxe.

A sintaxe da linguagem de programação Python é o `conjunto de regras` que define como um programa Python é escrito e interpretado.

A `clareza do código` deve ser válida tanto para o sistema de execução como para o ser humano.



## Shebang
Uma linha `shebang` é a primeira linha de um ficheiro de script e começa pelos caracteres cardinal e ponto de exclamação `#!`. Indica ao sistema operativo qual o interpretador a usar, por exemplo:

```python
#!/usr/bin/env python3
```

O `shebang` é muito útil em sistemas Linux e macOS, permitindo escolher a instalação de Python necessária ao projeto. No Windows, o lançador `py` também o respeita. Nos notebooks (Jupyter, VS Code, Colab) não é necessário.



## Identificadores
Os identificadores são os nomes dados a entidades como variáveis, funções, classes, etc., e ajudam a diferenciar uma entidade da outra.



## Regras
Os identificadores podem conter letras, dígitos e o sublinhado (*underscore*) `_`.

O nome pode começar por uma letra ou por um sublinhado `_` e pode ser complementado por qualquer número de letras, dígitos e sublinhados. É possível utilizar números no nome de um identificador, desde que não estejam no início do nome.

O nome de um identificador não pode conter caracteres especiais (como `@`, `#`, `-`) nem espaços. No decorrer do curso iremos ver vários casos, como por exemplo dar nome às variáveis.



## Indentação
A indentação (o recuo no início da linha) é obrigatória em Python, porque indica os blocos de código. Uma indentação errada provoca erros.

Algumas regras de indentação:
- Usar dois pontos `:` para iniciar um bloco e premir `[Enter]`;
- Todas as linhas do bloco devem usar a mesma indentação;
- Utilizar **quatro espaços** por nível de indentação e não misturar espaços com `[Tab]` no mesmo ficheiro.



## Comentários
É possível incluir comentários num programa Python. Isto ajuda a perceber o que foi feito quando o programa é lido por outra pessoa (ou por nós próprios, mais tarde).

Os comentários começam por `#` e vão até ao fim da linha. Executa a célula abaixo.


````python
# Este é um comentário: o Python ignora-o
print("Bem-vindo ao curso básico de Python")  # também pode estar no fim de uma linha
````

Para textos em várias linhas usam-se *strings* entre três aspas (`"""..."""`). Quando aparecem na primeira linha de um módulo, de uma função ou de uma classe, chamam-se **docstrings** e servem para documentar o código (veremos isto no capítulo de funções).


````python
"""
Este é um exemplo de texto
que abrange várias
linhas do programa!
"""
print("Procura sempre seguir boas práticas e documentar o teu código")
````

## Instruções (statements)
As instruções em Python terminam normalmente no fim da linha.

Quando uma instrução é muito longa, a forma recomendada de a continuar na linha seguinte é usar parênteses:


````python
total = (5 +
         5 +
         8 -
         3)
print(total)  # 15
````

Também é possível usar o carácter de continuação de linha `\`, mas é menos legível e desaconselhado.

O ponto e vírgula `;` permite escrever várias instruções numa única linha, desde que nenhuma inicie um novo bloco de código. Também é desaconselhado, mas convém conhecê-lo:


````python
x, y = 0, 1; z = x * y + x; print(z)
# 0 * 1 + 0 = 0
````

## Guia de estilo: PEP 8
A comunidade Python segue um guia de estilo oficial, a [PEP 8](https://peps.python.org/pep-0008/). As regras mais importantes para começar:

- 4 espaços por nível de indentação;
- nomes de variáveis e funções em minúsculas, com palavras separadas por `_` (por exemplo, `saldo_conta`, `calcular_media`);
- nomes de constantes em maiúsculas (por exemplo, `TAXA_IVA = 0.23`);
- um espaço de cada lado dos operadores de atribuição e de comparação (`x = 5`, `if x == 5:`);
- linhas com, no máximo, 79 caracteres;
- comentários claros, que expliquem o *porquê* e não apenas o *quê*.

Seguir a PEP 8 torna o código mais fácil de ler por qualquer programador Python.


## Palavras-chave
São as palavras reservadas pela linguagem Python e não podem ser utilizadas para nomear variáveis, funções ou qualquer outro identificador.

As `palavras-chave` definem a sintaxe e a estrutura da linguagem Python. Distinguem maiúsculas de minúsculas (*case sensitive*): `True` é uma palavra-chave, `true` não é.

É possível obter as `palavras-chave` diretamente do interpretador Python, com **duas linhas de código**:


````python
import keyword
print(keyword.kwlist)
````


O resultado da operação **acima** é uma `lista` (estrutura de dados que veremos com mais detalhes no decorrer do curso) com todas as palavras-chave da linguagem de programação Python.
