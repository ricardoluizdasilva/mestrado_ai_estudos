# Capítulo 11 – Ciclos (loops)

Em programação, laço (loop) ou estrutura de repetição, é uma instrução para o programa repetir tarefas num mesmo bloco de código.

Em Python existem 2 tipos de loops.





## For

É utilizado para iterar sobre uma sequência de dados como `string`, `range`, `list`, `set`, `tuple` e `dict`, executando um conjunto de instruções em cada elemento.
 
```Python
for <var> in <iterável>:
    <instruções>
    # break ou continue
else:  # opcional
    <instruções>
```

O bloco `else` é opcional e raramente usado: é executado quando o ciclo termina normalmente, sem ter sido interrompido por um `break` (há um exemplo mais abaixo).

No **exemplo abaixo**, o `for` itera sobre um `range()`.

````python
for i in range(10):
    print(i, end=' ')
````

### `Range`

A função `range()` permite gerar uma **sequência de números**. No `exemplo acima`, irá gerar números de `0 a 9 (10 números)`. Permite também definir o início da sequência, o fim (que não é incluído) e o passo. A sintaxe é `range(start,stop,step)`. Se o **step** não for indicado o **padrão é de 1 em 1**, conforme exemplificado anteriormente. No exemplo `abaixo` o **step** `será de 2 em 2`.

````python
for i in range(0, 55, 2):
    print(i, end = ' ')
    
````

No **próximo exemplo** o `for` itera sobre uma `list`.


````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for name in linux:
    print(name)
    
````


Neste **exemplo**, combinamos o `for` com uma condição.


````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for name in linux:
    print(name)
    if name == "suse":
        break
````

O `break` interrompe o ciclo. Se o ciclo terminar **sem** `break`, é executado o bloco `else` (se existir). No exemplo seguinte, experimenta procurar uma distribuição que não está na lista:


````python
procurar = "arch"
for name in linux:
    if name == procurar:
        print("Encontrada:", name)
        break
else:
    print(procurar, "não está na lista")
````

E que tal se o `for` iterar sobre um `dicionário (dict)`. Vejamos **este exemplo:**

````python
numeros = {'one': 1, 'two': 2, 'three': 3}   # evita chamar "dict" a uma variável: é o nome do tipo!
for chave in numeros:
    print(chave, numeros[chave])

# Alternativa: percorrer chave e valor ao mesmo tempo
for chave, valor in numeros.items():
    print(chave, "->", valor)
````

### `Enumerar valores`

A função `enumerate()` adiciona um **contador** a um iterável e devolve, em cada iteração, o par (contador, elemento). Recebe dois parâmetros:

- iterable: uma sequência ou qualquer objeto que suporte iteração;
- start (opcional): o número com que o contador começa; se for omitido, começa em 0.

Na **célula a seguir**, um exemplo:

````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for indice, distro in enumerate(linux):
    print(f'{indice} -> {distro}')
````

No `exemplo anterior` a enumeração é a mesma do índice de uma lista. No **próximo exemplo** o `contador começa em 1`.

````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for indice, distro in enumerate(linux, 1):
    print(f'{indice} -> {distro}')
````

**Sem o enumerate()**, a numeração do 1.º exemplo poderia ser feita assim:

````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
indice = 0 
for distr in linux:
    print(f'{indice} -> {distr}')
    indice += 1
````

Ou ainda desta **outra maneira**:

````python
linux = ["mandriva", "ubuntu", "fedora", "debian", "xubuntu", "suse", "red hat", "kali"]
for indice in range(len(linux)):
    print(f'{indice} -> {linux[indice]}')
````

### `Ciclos dentro de ciclos (aninhados)`

Um ciclo pode conter outro ciclo. Por cada iteração do ciclo exterior, o ciclo interior é executado do início ao fim. Exemplo: a tabuada do 1 ao 3.


````python
for i in range(1, 4):
    for j in range(1, 11):
        print(f"{i} x {j:2} = {i * j:2}", end="   ")
    print()  # muda de linha no fim de cada tabuada
````

## While

É utilizado quando não sabemos à partida quantas iterações vão ocorrer. Repete um bloco de instruções **enquanto** a condição for verdadeira (`True`) e termina quando a condição passar a ser `False`.

```Python
while <expressão>:
    <instruções>
    # break ou continue
```



> **Atenção:** se a condição nunca passar a `False`, o ciclo é infinito. Para o interromper, usa o botão ■ (*Interromper*) no Jupyter e no VS Code, ou o botão de parar ao lado da célula no Colab.

Abaixo, um **exemplo** de `while` em que uma variável utiliza um `contador` para incrementar o seu valor.


````python
x = 0
while x < 10:
    print(x, end=" ")
    x += 1 # x = x + 1
````


Um outro **exemplo** é a utilização de um boolean (`True`). Como a condição é sempre verdadeira, o ciclo só termina quando encontra um `break`.


````python
x = 0
while True:
    print(x, end=" ")
    x += 1 # x = x + 1
    if x == 5:
        break
````

Neste **outro exemplo** `subtraímos um do contador`. E para tal invertemos o `operador da condição` e o seu `valor de comparação`.


````python
x = 10
while x > 0:
    print("x = ", x)
    x -= 1 # x = x - 1
````


Vejamos agora um **exemplo** com `continue`.


````python
x = 10
while x > 0:
    x -= 1
    if x == 5:
        continue
    print(x, end=" ")
````

## For vs While

O `for` é utilizado nas seguintes situações:
- Percorrer os elementos de uma sequência (lista, string, `range`, etc.)
- O número de iterações é conhecido à partida
- Pode terminar antecipadamente com o `break`


O `while` é utilizado quando:
- O número de iterações é desconhecido à partida (depende de uma condição)
- Pode terminar antecipadamente com o `break`
- Se for usado um contador, é necessário inicializá-lo antes do ciclo e atualizá-lo dentro do bloco de código.
