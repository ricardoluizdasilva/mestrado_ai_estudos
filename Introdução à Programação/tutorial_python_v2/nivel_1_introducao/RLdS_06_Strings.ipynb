# Capítulo 6 – Strings


## Definição

É um tipo de dados que representa uma `sequência de caracteres`, que podem ser letras, dígitos, símbolos ($, !, #, @), etc.

Um **carácter** é simplesmente um `símbolo`. Por exemplo, o alfabeto do idioma português possui 26 caracteres.

Os computadores lidam fundamentalmente com números `binários`. No ecrã estaremos a ver o carácter, mas internamente eles são armazenados e manipulados como uma combinação de `0 e 1`

A conversão de caracteres em números é chamada de `codificação` e o processo inverso é a `descodificação`. `ASCII` e `UNICODE` são algumas das codificações mais usadas. Em `Python`, uma `string` é uma sequência de `caracteres Unicode`, que foi introduzido para incluir todos os caracteres de todos os idiomas, uniformizando a codificação.





A função `ord()` devolve o código numérico Unicode (*code point*) de um carácter. Para os caracteres da tabela ASCII (0 a 127), esse código coincide com o valor ASCII.


````python
print(ord('A'))
print(ord('<'))
print(ord('∑'))
print(ord('火'))
````

A função `chr()` faz o `inverso` de `ord()`, ou seja, a partir do `valor numérico` devolve uma `string` com o carácter correspondente.


````python
print(chr(65))
print(chr(60))
print(chr(8721))
print(chr(28779))
print(chr(960))
````

### `Índices`

Em Python, uma `string` é uma sequência de caracteres Unicode. Não existe um tipo de dados próprio para um carácter: um único carácter é uma string de `tamanho 1`.

A função `len()` (uma função *built-in*) retorna o tamanho (comprimento) de um objeto. No caso das strings, retorna o número de caracteres.



````python
msg = "PROGRAMADORES"
print(msg, "tem", len(msg), "caracteres")
````

Saber quantos caracteres tem uma string permite determinar quais são os seus índices.



O índice de uma string (lista, tuplo, etc.) inicia na posição 0. Os *índices negativos* contam a partir do fim: o último carácter tem o índice `-1`, o penúltimo `-2` e o primeiro `-len(string)`, conforme exemplificado acima.

Para aceder a uma posição de uma string basta utilizar a string seguida `[]` com o número que indica o `índice`. A sintaxe é `string[n]`.

```python
a = "Olá"
print(a[1]) # retorna l
```

Isto permite `iterar` sobre uma `string`, ou extrair partes dela (`slicing`), como se vê a seguir; mais detalhes no capítulo de `listas`.

Se a string tem uma única posição, o seu índice será `0` ou `-1`.

```python
msg = "a"
print(msg[0]) # ou -1 --> retorna a
```

````python
msg = "PROGRAMADORES"
print(msg[1], msg[6], msg[-6])
````

### `Fatias (slicing), concatenação e imutabilidade`

- `string[início:fim]` devolve a parte da string do índice `início` (incluído) até ao índice `fim` (excluído).
- O operador `+` junta (concatena) strings e o operador `*` repete-as.
- O operador `in` verifica se um texto existe dentro de outro.
- As strings são **imutáveis**: não é possível alterar um carácter (`msg[0] = "p"` dá erro). Para "alterar" uma string, cria-se uma nova.


````python
msg = "PROGRAMADORES"
print(msg[0:7])      # PROGRAM
print(msg[7:])       # ADORES
print(msg[-3:])      # RES
print("Olá " + "mundo")
print("ha" * 3)
print("GRAMA" in msg)
nova = "p" + msg[1:].lower()   # cria uma nova string
print(nova)
````

### `Métodos`

Existem diversos métodos (funções especiais) que permitem manipular strings.

- `strip():` remove todos os espaços extras no início e fim da string.    
- `lower():` retorna a string em lowercase (minúsculas).
- `upper():` retorna a string em uppercase (maiúsculas).
- `swapcase():` retorna a string com caracteres maiúsculos convertidos em minúsculos e vice-versa.
- `title():` retorna a string no qual a primeira letra de cada palavra é convertida para maiúscula e as demais letras permanecem em minúsculas.
- `replace():` substitui uma string por outra string.
- `split():` separa a string em substrings caso haja um separador.
- `join():` retorna a string que resulta da concatenação dos elementos de um iterável separados por um delimitador.
- `count():` retorna o número de ocorrências não sobrepostas de uma substring.
- `startswith():` retorna `True` se uma string inicia com um prefixo especificado.
- `endswith():` retorna `True` se uma determinada string termina com um sufixo especificado.
- `find():` pode ser utilizado para ver se uma string contém uma substring informada e retorna o menor índice na string onde esta substring é encontrada.
- `isalnum():` retorna `True` se a string for `não-vazia` e todos os seus caracteres forem `alfanuméricos` (ou uma letra ou um número).
- `isalpha():` retorna `True` se a string for `não-vazia` e todos os seus caracteres forem `alfabéticos`.
- `isdigit():` retorna `True` se a string for `não-vazia` e contém apenas `dígitos`.
- `isidentifier():` retorna `True` se uma determinada string é um identificador válido de acordo com a definição da linguagem Python.
- `iskeyword()`: não é um método de string, mas uma função do módulo `keyword`, que testa se uma string é uma `palavra-chave` do Python. Tem de ser importada antes (`from keyword import iskeyword`).
- `isprintable():` retorna `True` se a string for vazia ou se todos os seus caracteres forem imprimíveis; caracteres de controlo como `\t` (tabulação) ou `\n` (nova linha) não são imprimíveis.
- `isspace():` determina se uma string consiste de caracteres de `espaço em branco`. Retornará `True` se a string for não-vazia e todos os caracteres forem de espaço em branco.
- `isupper():` verifica se todos os caracteres de uma string são maiúsculos.
- `islower():` verifica se todos os caracteres de uma string são minúsculos.
- `center():` retorna uma string centrada num determinado comprimento especificado via argumento. Por definição padrão, o preenchimento consiste no carácter de `espaço` ASCII.
- `expandtabs():` substitui cada carácter tab `(\t)` com espaços.
- `ljust():` retorna uma string justificada à esquerda num campo de comprimento especificado via argumento. O preenchimento padrão consiste no carácter de `espaço` ASCII.
- `rjust():` retorna uma string justificada à direita num campo de comprimento especificado via argumento. Por definição, o preenchimento padrão consiste no carácter `espaço` ASCII.
- `zfill():` retorna uma cópia da string com preenchimento à esquerda com caracteres '0' e o comprimento é especificado como argumento.
- `format():` permite construir strings de uma forma flexível. *Iremos abordar mais detalhes no capítulo específico de formatação de saída de dados.

Todos os métodos *`.isalgo()`* retornam valores boolean (`True` ou `False`).

````python
# Importar a função iskeyword e definir variáveis
from keyword import iskeyword  # os módulos serão abordados no capítulo 21
msg = " Olá aulnos do curso de básico de Python"   # "aulnos" está errado de propósito (ver replace)
alunos = ["Maria", "João", "José", "Nelson", "Beatriz", "Isabel"]
nome = alunos[0]
profissao = "estudante"

print(msg.strip())
print(msg.lower())
print(msg.upper())
print(msg.swapcase())
print(msg.title())
print(msg.replace("aulnos", "alunos"))
print(msg.split(" "))
print(', '.join(alunos))
print(' - '.join(alunos))
print(list('aeiou'))
print('|'.join('aeiou'))
print("Casa Cadeira Lâmpada casa Madeira Telhado Casa".count("Casa"))
print("parlamentarismo".startswith("parla"))
print("parlamentarismo".startswith("ismo"))
print("parlamentarismo".endswith("parla"))
print("parlamentarismo".endswith("ismo"))
print("Estudar muito, é o lema dos Programadores".find("muito"))
print("Estudar muito, é o lema dos Programadores".find("Estudar"))
print("abc123".isalnum())
print("abx#123".isalnum())
print("exemplo".isalpha())
print("exemplo 2".isalpha())
print("49".isdigit())
print("49c".isdigit())
print("".isdigit())
print("nome".isidentifier())
print("2nome".isidentifier())
print("nome#".isidentifier())
print(iskeyword("and"))
print(iskeyword("switch"))
print(iskeyword("else"))
print(iskeyword("while"))
print("Darth\tVader".isprintable())
print("Darth Vader".isprintable())
print("".isprintable())
print("x\ny".isprintable())
print("".isspace())
print("x".isspace())
print(" ".isspace())
print("\t\n".isspace())
print("ALUNO".isupper())
print("Aluno".isupper())
print("Aluno".islower())
print("aluno".islower())
print(nome.center(20))
print(nome.center(25,"."))
print("x\ty".expandtabs())
print("C++".ljust(10))
print("C++".ljust(10,"-"))
print("Javascript".rjust(20))
print("JavaScript".rjust(20,"#"))
print("66".zfill(4))
print(nome.zfill(7))
print("A profissão de {0} é {1}".format(nome, profissao))
````

### `Programa para imprimir no ecrã a tabela ASCII completa (0 a 127)`

Este programa usa funções, ciclos e f-strings, que só veremos mais à frente. Por agora, executa-o e observa o resultado; no fim do curso, volta aqui e tenta perceber cada linha.


````python

control_names = [
    "NUL", "SOH", "STX", "ETX", "EOT", "ENQ", "ACK", "BEL",
    "BS", "HT", "LF", "VT", "FF", "CR", "SO", "SI",
    "DLE", "DC1", "DC2", "DC3", "DC4", "NAK", "SYN", "ETB",
    "CAN", "EM", "SUB", "ESC", "FS", "GS", "RS", "US"
]

def ascii_char(n):
    if n < 32:
        return control_names[n]
    if n == 32:
        return "SP"
    if n == 127:
        return "DEL"
    return chr(n)

header = f"{'Dec':<4} {'Hx':<3}  {'Oct':<3}  {'Char':<5}"
print(" | ".join([header] * 4))
print("-" * 107)

for row in range(32):
    columns = []

    for group in range(4):
        n = row + group * 32
        columns.append(
            f"{n:<4} {n:02X}  {n:03o}  {ascii_char(n):<5}"
        )

    print(" | ".join(columns))
````
