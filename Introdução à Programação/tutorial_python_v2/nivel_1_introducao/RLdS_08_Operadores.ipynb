# Capítulo 8 – Operadores

São símbolos especiais usados na construção de expressões, as quais contêm um número variado de operandos. Por exemplo, a expressão `a + b`, temos o operador `+` e os operandos `a` e `b`.

Além das operações aritméticas, em Python é possível fazer outras operações, tais como **`atribuição, comparação, lógicas, identidade, pertença (membro) e binária (bitwise)`**.

Conhecer estas operações é importante porque em programação são utilizadas praticamente a toda a hora no **controlo de fluxo e na tomada de decisões**.

## Precedência

Quem se recorda do **PEMDAS**?

Todos os operadores em Python têm precedência. Em geral, seguem as regras da matemática.

Numa expressão, todos os operadores de maior precedência são executados primeiro. Depois, são executados os operadores da próxima precedência mais alta. E isto continua até que a expressão seja totalmente avaliada. Os operadores de igual precedência são executados da esquerda para a direita.

A tabela a seguir apresenta a ordem de precedência dos operadores, do menor para o maior:





Os operadores da parte superior têm a precedência mais baixa. Os localizados na parte inferior têm a precedência mais alta. Quaisquer operadores na mesma linha da tabela têm igual precedência.

A **precedência do operador** pode ser substituída usando os parênteses `()`. As expressões entre parênteses são sempre executadas primeiro, antes das expressões que não estão entre parênteses.

**Exemplos:**


````python
print(5 + 5 * 5)    # resultado 30
print((5 + 5) * 5)  # resultado 50
````

## Operadores relacionais

| Símbolo   | Descrição        | Exemplo                |Significado                       |
| --------- | ---------------- | ---------------------- |--------------------------------- |
| >         | Maior que        | X > Y                  |`X` é maior que `Y` ?             |

Os operadores relacionais são utilizados para comparar valores (de qualquer tipo). O resultado de uma expressão relacional será um resultado `booleano` (True ou False).

## Operadores lógicos

Em Python existem apenas três operadores lógicos: `and`, `or` e `not`. Os restantes (NAND, NOR, XOR e XNOR) obtêm-se combinando estes três, como se mostra a seguir.

### **`Operador AND`**

| VALOR 1          | VALOR 2          | AND (RESULTADO)  |
| ---------------- | ---------------- | ---------------- |
| `VERDADEIRO`     | `VERDADEIRO`     | `VERDADEIRO`     |
| VERDADEIRO       | FALSO            | FALSO            |
| FALSO            | VERDADEIRO       | FALSO            |
| FALSO            | FALSO            | FALSO            |

O resultado será `VERDADEIRO` se os **dois** valores de entrada da operação forem `VERDADEIRO`.

````python
x = 0
y = 1
print(x < y and 0 == x)
````

### **`Operador OR`**

| VALOR 1          | VALOR 2          | OR (RESULTADO)   |
| ---------------- | ---------------- | ---------------- |
| `VERDADEIRO`     | `VERDADEIRO`     | `VERDADEIRO`     |
| `VERDADEIRO`     | FALSO            | `VERDADEIRO`     |
| FALSO            | `VERDADEIRO`     | `VERDADEIRO`     |
| FALSO            | FALSO            | FALSO            |

O resultado será `VERDADEIRO` se **um dos dois valores** de entrada da operação for `VERDADEIRO`.


````python
x = 0
y = 1
print(x > y or 0 == x)
````

### **`Operador NAND`**

| VALOR 1          | VALOR 2          | NAND (RESULTADO) |
| ---------------- | ---------------- | ---------------- |
| VERDADEIRO       | VERDADEIRO       | FALSO            |
| `VERDADEIRO`     | `FALSO`          | `VERDADEIRO`     |
| `FALSO`          | `VERDADEIRO`     | `VERDADEIRO`     |
| `FALSO`          | `FALSO`          | `VERDADEIRO`     |

Este operador é o **oposto** do `AND`, ou seja, resulta em `VERDADEIRO` se ao menos um dos valores de entrada da operação for `FALSO`.

````python
x = 0
y = 1
print(not(x < y and 0 == x))
````

### **`Operador NOR`**

| VALOR 1          | VALOR 2          | NOR (RESULTADO)  |
| ---------------- | ---------------- | ---------------- |
| VERDADEIRO       | VERDADEIRO       | FALSO            |
| VERDADEIRO       | FALSO            | FALSO            |
| FALSO            | VERDADEIRO       | FALSO            |
| `FALSO`          | `FALSO`          | `VERDADEIRO`     |

Este operador é o **oposto** do `OR`, ou seja, resulta em `VERDADEIRO` se os dois valores de entrada da operação forem `FALSO`.

````python
x = 0
y = 1
print(not(x == y or 1 == x))
````

### **`Operador XOR`**

| VALOR 1          | VALOR 2          | XOR (RESULTADO)  |
| ---------------- | ---------------- | ---------------- |
| VERDADEIRO       | VERDADEIRO       | FALSO            |
| `VERDADEIRO`     | `FALSO`          | `VERDADEIRO`     |
| `FALSO`          | `VERDADEIRO`     | `VERDADEIRO`     |
| FALSO            | FALSO            | FALSO            |

O operador `XOR` é uma variação do operador `OR`, e resulta em `VERDADEIRO` se apenas um dos valores de entrada for verdadeiro, ou seja, se os valores de entrada forem diferentes.


````python
x = False
y = True
print((x and not y) or (not x and y))
print(x != y)  # com booleanos, XOR equivale a "diferente"
````

### **`Operador XNOR`**

| VALOR 1          | VALOR 2          | XNOR (RESULTADO) |
| ---------------- | ---------------- | ---------------- |
| `VERDADEIRO`     | `VERDADEIRO`     | `VERDADEIRO`     |
| VERDADEIRO       | FALSO            | FALSO            |
| FALSO            | VERDADEIRO       | FALSO            |
| `FALSO`          | `FALSO`          | `VERDADEIRO`     |

O operador `XNOR` é o contrário do `XOR`, e resulta em `VERDADEIRO` se os valores de entrada forem iguais.

````python
x = True
y = True
print((x or not y) and (not x or y))
print(x == y)  # com booleanos, XNOR equivale a "igual"
````

### **`Operador NOT`**

| VALOR ENTRADA    | NOT (RESULTADO)  |
| ---------------- | ---------------- |
| `VERDADEIRO`     | FALSO            |
| FALSO            | `VERDADEIRO`     |

É o único operador que recebe como entrada apenas um valor. A sua função é inverter o valor de entrada.

````python
x = 0
y = 1
print(not(x == y))
````

### `Valores verdadeiros e falsos e avaliação em curto-circuito`

Em Python, qualquer valor pode ser usado como condição. São considerados **falsos**: `False`, `0`, `0.0`, `""` (string vazia), `[]`, `{}`, `()` e `None`. Todos os outros valores são considerados **verdadeiros**.

Além disso, `and` e `or` não devolvem necessariamente `True` ou `False`, mas sim **um dos operandos**, e param de avaliar assim que o resultado fica decidido (*curto-circuito*):
- `a or b` devolve `a` se `a` for verdadeiro; caso contrário devolve `b`;
- `a and b` devolve `a` se `a` for falso; caso contrário devolve `b`.


````python
print(0 or 1)          # 1 (e não True)
print(0 and 1)         # 0
print("" or "vazio")   # vazio
print(bool(0 or 1))    # True
````

## Operador de pertença
Utiliza-se o operador `in` (de pertença ou membro) para verificar se um determinado valor é um elemento que pertence a uma variável (string, array, lista, tuplo, etc.) ou valor. O resultado da operação será um resultado `booleano` (Verdadeiro ou Falso).

Pode ser combinado com o operador `not`. Neste caso o resultado da operação será verdadeiro se o elemento não estiver presente na operação.

**Exemplo:**


````python
msg = "O instrutor é o Ricardo"
print("Ricardo" in msg)

nlist = [0, 1, 2 , 3, 4, 5, 6, 7, 8]
print(9 not in nlist)
````

## Operador de identidade
O operador `is` verifica se duas variáveis se referem ao **mesmo objeto** em memória (identidade), e não apenas se têm o mesmo valor (para isso usa-se `==`). O resultado é um `booleano` (`True` ou `False`).

Pode ser combinado com o operador `not`: `is not` devolve `True` quando os objetos são diferentes.

Abaixo, um **exemplo** com números.

````python
x = 1
y = 1
print(x is y)
print(x is not y)
````

**Exemplo de identidade com string:**

````python
nome = "Ricardo"
apelido = "Ricardo"
print(nome is apelido)
print(nome is not apelido)
````

**No caso de listas**, tem que se observar que o interpretador Python não as considera idênticas, isto porque o interpretador localiza os objetos separadamente em memória, mesmo sendo eles iguais.


````python
x = [1, 2, 3]
y = [1, 2, 3]
print(x is y)
print(x is not y)
````

> **Atenção:** nos exemplos com números pequenos e com strings curtas, `is` deu `True` porque o Python reutiliza internamente esses objetos, como otimização. Não se deve contar com isso: **para comparar valores usa-se sempre `==`**. O `is` usa-se sobretudo para comparar com `None` (por exemplo, `if resultado is None:`).


## Operadores Bitwise
Os operadores Bitwise atuam nos operandos ao nível binário (0 e 1), como se fossem strings ou dígitos binários, executando as operações bit a bit. Ou seja, fazer operações lógicas com os bits de dois números.

As operações bitwise são utilizadas quando é necessário realizar operações na representação binária de números inteiros.

**Nota:** a função `bin()` retorna a string binária de um número:

```python
x = 9
y = 7
print(bin(x)) # 0b1001
print(bin(y)) # 0b0111
```

Vamos considerar o valor `9` que é representado por `1001` em binário e `7` que é representado por `0111` para realizarmos as operações binárias.

Para as operações binárias utilizamos os operadores lógicos de bitwise. Vejamos os exemplos abaixo:



Vamos analisar dois casos, o `AND` e o `OR`.

```python
x = 0b1001 # representação binária do 9
y = 0b0111 # representação binária do 7

z = x & y  # retorna 1 - representação binária 0b0001
# x = 0b1001
# y = 0b0111
# ----------
# z = 0b0001

z = x | y  # retorna 15 - representação binária 0b1111
# x = 0b1001
# y = 0b0111
# ----------
# z = 0b1111
```

````python
x = 9
y = 7
z = x  & y
print("A operação de bitwise AND retorna", z)
print("A sua representação binária é", bin(z),"\n")
z = x | y
print("A operação de bitwise OR retorna", z)
print("A sua representação binária é", bin(z),"\n")
````

### `Sequence types (Tipos de sequência)`

Em Python há três tipos básicos de tipos de sequência (`sequence types`), que são `list`, `range` e `tuple`. 
Um outro tipo de sequência bastante comum é `string`, usado exclusivamente para textos, que pode conter caracteres Unicode. Durante o curso iremos abordar esta e outras `sequence types`.

As sequências têm em comum:
- Consistem em vários itens, que estão ordenados (não significa que estejam organizados por tamanho, por exemplo. Indica simplesmente que existe um primeiro elemento, um segundo elemento, etc.)

- Pode ser utilizado um índice para aceder a um elemento específico na sequência.


É importante mencionar algumas operações que são comuns entre as `sequence types`, conforme indicado na tabela abaixo:

| Operação  | Valor de retorno
| --------- | --------------------------------------------------------------------- |
| `len(s)`  | Tamanho da sequência                                                  |
| `min(s)`  | Menor elemento da sequência                                           |
| `max(s)`  | Maior elemento da sequência                                           |
| `s + t`   | Concatenação de `s` e `t`                                             |
| `i * s`   | Uma nova sequência que repete `s` `i` vezes                           |
| `s[i]`    | Aceder ao elemento de índice `i` da sequência `s`                      |
| `i in s`  | Verificar se a sequência `s` contém o elemento  `i`                   |


No **capítulo 6** foi mencionado o termo slicing, algo que será abordado no **capítulo 12**, onde também iremos abordar o acesso a um elemento da lista pelo seu índice.

Nota importante: nem todas as operações irão funcionar para todas as sequências. Por exemplo, um `range` não pode ser multiplicado por um inteiro para criar um range maior. Na leitura recomendada estão as fontes de informação onde poderão obter mais informações sobre todas as sequências, incluindo as limitações. 

No **capítulo 6**, vimos em detalhes as `strings`, que são imutáveis, ou seja, uma string não pode ser modificada depois de criada.

Para alterar uma string deve-se primeiro construir uma nova string através da concatenação ou função, para então atribuir novamente à variável string.

Para as operações com strings usamos operadores especiais, sendo que já vimos alguns deles (`in` e `not in`).

- `+`: concatenação, adiciona os valores
- `*`: repetição, concatena várias cópias da mesma string
- `in`: verifica se um ou mais caracteres existem na string
- `not in`: verifica se um ou mais caracteres não existem na string

**Exemplos:**


````python
nome = "Darth"
apelido = "Vader"
print(nome + " " + apelido)
print((nome + " ") * 10)
print("a" in nome)
print("w" not in apelido)
````
